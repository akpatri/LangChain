# Global Configuration API Reference

Developer-facing statements defined in `libs/core/langchain_core/globals.py`.

# `set_verbose`

Sets the global verbose setting, replacing its previous value.

```python
set_verbose(
    value: bool, # New value for the verbose setting
) -> None
```

---

# `get_verbose`

Returns the current global verbose setting.

```python
get_verbose(
) -> bool # Current verbose setting
```

---

# `set_debug`

Sets the global debug setting, replacing its previous value.

```python
set_debug(
    value: bool, # New value for the debug setting
) -> None
```

---

# `get_debug`

Returns the current global debug setting.

```python
get_debug(
) -> bool # Current debug setting
```

---

# `set_llm_cache`

Sets the global LLM cache, replacing any previously configured cache. Passing `None` disables the LLM cache.

`BaseCache` is imported only under `TYPE_CHECKING`; the annotation is stored as a forward reference.

```python
set_llm_cache(
    value: Optional["BaseCache"], # New LLM cache, or None to disable caching
) -> None
```

---

# `get_llm_cache`

Returns the currently configured global LLM cache.

`BaseCache` is imported only under `TYPE_CHECKING`; the annotation is stored as a forward reference.

```python
get_llm_cache(
) -> Optional["BaseCache"] # Configured LLM cache, or None
```

````python
from langchain_core.caches import InMemoryCache # Import LangChain's in-memory LLM cache
from langchain_core.globals import get_debug # Import the global debug-setting getter
from langchain_core.globals import get_llm_cache # Import the global LLM-cache getter
from langchain_core.globals import get_verbose # Import the global verbose-setting getter
from langchain_core.globals import set_debug # Import the global debug-setting setter
from langchain_core.globals import set_llm_cache # Import the global LLM-cache setter
from langchain_core.globals import set_verbose # Import the global verbose-setting setter

print("Initial verbose:", get_verbose()) # Display the initial global verbose setting
print("Initial debug:", get_debug()) # Display the initial global debug setting
print("Initial cache:", get_llm_cache()) # Display the initially configured LLM cache

set_verbose(True) # Enable verbose output globally
set_debug(True) # Enable debug output globally

cache = InMemoryCache() # Create a real in-memory LangChain cache
set_llm_cache(cache) # Configure the cache as the global LLM cache

print("Updated verbose:", get_verbose()) # Confirm that verbose mode is enabled
print("Updated debug:", get_debug()) # Confirm that debug mode is enabled
print("Configured cache:", type(get_llm_cache()).__name__) # Display the configured cache class name

set_verbose(False) # Disable verbose output globally
set_debug(False) # Disable debug output globally
set_llm_cache(None) # Disable the global LLM cache

print("Final verbose:", get_verbose()) # Confirm that verbose mode is disabled
print("Final debug:", get_debug()) # Confirm that debug mode is disabled
print("Final cache:", get_llm_cache()) # Confirm that no global LLM cache is configured
````
