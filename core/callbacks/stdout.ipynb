# `StdOutCallbackHandler`

Callback handler that prints chain, agent, tool, and text callback output to standard output.

It uses `print` for chain lifecycle messages and `print_text` for agent logs, tool output, and arbitrary text.

- Bases: `BaseCallbackHandler`

## Constructor

```python
StdOutCallbackHandler(
    color: str | None = None # Default colour used by print_text
)
```

## Parameters

* `color` — Default colour used when printing agent actions, tool output, text, and agent completion logs.
  * Default: `None`
  * A colour passed directly to a callback method takes precedence over this default.

## Attributes

* `color` — Stores the default output colour.

## Methods

1. `on_chain_start`: Prints a bold message when a chain begins.
   * Determines the displayed chain name in this order:
     1. `kwargs["name"]`, when provided.
     2. `serialized["name"]`, when present.
     3. The final value in `serialized["id"]`.
     4. `"<unknown>"`.
   * The `inputs` argument is accepted but is not printed.
   - **Syntax:**
     ```python
     on_chain_start(
         self,
         serialized: dict[str, Any], # Serialized chain information
         inputs: dict[str, Any], # Inputs passed to the chain
         **kwargs: Any # May contain an explicit chain name
     ) -> None
     ```
   - **Output format:**
     ```text
     > Entering new <name> chain...
     ```

2. `on_chain_end`: Prints a bold message when a chain finishes.
   * The `outputs` argument is accepted but is not printed.
   - **Syntax:**
     ```python
     on_chain_end(
         self,
         outputs: dict[str, Any], # Outputs produced by the chain
         **kwargs: Any # Additional callback arguments
     ) -> None
     ```
   - **Output:**
     ```text
     > Finished chain.
     ```

3. `on_agent_action`: Prints the log associated with an agent action.
   * Uses the method-level `color` when supplied.
   * Otherwise uses the handler's default `color`.
   - **Syntax:**
     ```python
     on_agent_action(
         self,
         action: AgentAction, # Agent action containing the log
         color: str | None = None, # Colour for this output
         **kwargs: Any # Additional callback arguments
     ) -> Any
     ```
   - **Printed value:**
     ```python
     action.log
     ```

4. `on_tool_end`: Prints the output returned by a tool.
   * Converts `output` to a string before printing.
   * Optionally prints an observation prefix before the output.
   * Optionally prints an LLM prefix after the output.
   * Uses the method-level `color` when supplied; otherwise uses the default handler colour.
   - **Syntax:**
     ```python
     on_tool_end(
         self,
         output: Any, # Tool result to print
         color: str | None = None, # Colour for the tool output
         observation_prefix: str | None = None, # Text printed before output
         llm_prefix: str | None = None, # Text printed after output
         **kwargs: Any # Additional callback arguments
     ) -> None
     ```
   - **Printing order:**
     ```text
     <observation_prefix>
     <tool output>
     <llm_prefix>
     ```
   - Prefixes are printed only when they are not `None`.

5. `on_text`: Prints arbitrary or intermediate text.
   * Uses the method-level `color` when supplied.
   * Otherwise uses the handler's default `color`.
   * The `end` argument controls what is appended after the text.
   - **Syntax:**
     ```python
     on_text(
         self,
         text: str, # Text to print
         color: str | None = None, # Colour for this output
         end: str = "", # Text appended after the output
         **kwargs: Any # Additional callback arguments
     ) -> None
     ```

6. `on_agent_finish`: Prints the final agent log.
   * Uses the method-level `color` when supplied.
   * Otherwise uses the handler's default `color`.
   * Appends a newline after the log.
   - **Syntax:**
     ```python
     on_agent_finish(
         self,
         finish: AgentFinish, # Agent finish object containing the final log
         color: str | None = None, # Colour for this output
         **kwargs: Any # Additional callback arguments
     ) -> None
     ```
   - **Printed value:**
     ```python
     finish.log
     ```

## Colour Precedence

Methods that support coloured output use:

```python
color or self.color
```

Therefore:

```text
Method-level colour -> used first
Handler default      -> used when method colour is absent
None                 -> no configured colour
```

## Example

```python
from langchain_core.callbacks.stdout import StdOutCallbackHandler

handler = StdOutCallbackHandler(
    color="green"
)

handler.on_text(
    "Processing..."
)
```
