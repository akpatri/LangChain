# Standard Output Callback Handler Reference

Developer-facing statements defined in `langchain_core.callbacks.stdout`.

# `StdOutCallbackHandler: BaseCallbackHandler`

Callback handler that prints chain, agent, tool, and text events to standard output.

## Fields

```python
color: str | None # Default color used when an event-specific color is not provided
```

## Constructor

```python
StdOutCallbackHandler(
    color: str | None = None, # Default color for printed text
) -> None
```

Initializes the callback handler with an optional default text color.

## Methods

### `on_chain_start`

```python
on_chain_start(
    serialized: dict[str, Any], # Serialized chain metadata used to determine the displayed chain name
    inputs: dict[str, Any], # Inputs supplied to the chain
    **kwargs: Any, # Additional callback data; `name` overrides the serialized chain name when present
) -> None
```

Prints a bold message indicating that a chain is starting. Uses `<unknown>` when no chain name can be determined.

### `on_chain_end`

```python
on_chain_end(
    outputs: dict[str, Any], # Outputs produced by the chain
    **kwargs: Any, # Additional callback data
) -> None
```

Prints a bold message indicating that the chain has finished.

### `on_agent_action`

```python
on_agent_action(
    action: AgentAction, # Agent action whose log is printed
    color: str | None = None, # Event-specific color overriding the handler default when truthy
    **kwargs: Any, # Additional callback data
) -> Any
```

Prints the agent action log.

### `on_tool_end`

```python
on_tool_end(
    output: Any, # Tool output converted to text before printing
    color: str | None = None, # Event-specific color overriding the handler default when truthy
    observation_prefix: str | None = None, # Optional text printed before the tool output
    llm_prefix: str | None = None, # Optional text printed after the tool output
    **kwargs: Any, # Additional callback data
) -> None
```

Prints the tool output, optionally surrounded by observation and LLM prefixes.

### `on_text`

```python
on_text(
    text: str, # Text to print
    color: str | None = None, # Event-specific color overriding the handler default when truthy
    end: str = "", # String appended after the printed text
    **kwargs: Any, # Additional callback data
) -> None
```

Prints callback text using the selected color and ending string.

### `on_agent_finish`

```python
on_agent_finish(
    finish: AgentFinish, # Agent finish result whose log is printed
    color: str | None = None, # Event-specific color overriding the handler default when truthy
    **kwargs: Any, # Additional callback data
) -> None
```

Prints the final agent log followed by a newline.

## Behaviour

Event-specific colors take precedence when truthy; otherwise, the handler-level `color` is used. Chain boundary messages are written directly to standard output, while agent, tool, and text content is emitted through `print_text`.

````python
# Install once in Jupyter Notebook if langchain-core is not already installed.
# !pip install langchain-core

from langchain_core.callbacks.stdout import StdOutCallbackHandler
from langchain_core.runnables import RunnableLambda


def create_greeting(name: str) -> str:
    """Clean the given name and create a greeting message."""
    cleaned_name = name.strip().title()
    return f"Hello, {cleaned_name}!"


# Convert the Python function into a LangChain runnable.
greeting_chain = RunnableLambda(create_greeting).with_config(
    {"run_name": "GreetingChain"}
)

# Create a callback handler that prints chain execution information.
stdout_handler = StdOutCallbackHandler(color="green")

# Take input from the user.
user_name = input("Enter your name: ")

# Execute the runnable with the callback handler.
result = greeting_chain.invoke(
    user_name,
    config={"callbacks": [stdout_handler]},
)

# Display the final result.
print("\nFinal Result:")
print(result)
````
