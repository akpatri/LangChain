# Input Formatting Utilities API Reference

Developer-facing statements defined in `libs/core/langchain_core/utils/input.py`.

# `get_color_mapping`

Assigns supported terminal colors to items in sequence.

```python
get_color_mapping(
    items: list[str], # Items to assign colors to
    excluded_colors: list[str] | None = None, # Supported color names to exclude
) -> dict[str, str] # Mapping from each item to a color name
```

Available colors are considered in this order: `"blue"`, `"yellow"`, `"pink"`, `"green"`, and `"red"`. Exclusions preserve the order of the remaining colors.

Colors are assigned cyclically, so the sequence repeats when there are more items than available colors. Duplicate item names overwrite earlier entries in the returned dictionary.

Raises `ValueError` when exclusions leave no available colors.

---

# `get_colored_text`

Wraps text in ANSI escape sequences for the selected supported color, with bold and italic styling.

```python
get_colored_text(
    text: str, # Text to format
    color: str, # Supported color name
) -> str # ANSI-formatted text
```

Raises `KeyError` when `color` is not one of `"blue"`, `"yellow"`, `"pink"`, `"green"`, or `"red"`.

---

# `get_bolded_text`

Wraps text in ANSI escape sequences for bold styling.

```python
get_bolded_text(
    text: str, # Text to format
) -> str # ANSI-formatted bold text
```

---

# `print_text`

Prints text with an optional supported terminal color.

```python
print_text(
    text: str, # Text to print or write
    color: str | None = None, # Supported color name, or None for unformatted text
    end: str = "", # String appended after the text
    file: TextIO | None = None, # Text stream to write to instead of standard output
) -> None
```

When `color` is provided, the function formats the text through `get_colored_text`. When `file` is provided, the stream is flushed after writing.

Raises `KeyError` through `get_colored_text` when `color` is unsupported.

````python
# 1. Generate a color mapping
from langchain_core.utils.input import get_bolded_text, get_color_mapping, get_colored_text, print_text # Import the text-formatting utilities


items = ["Search", "Retriever", "LLM", "Parser", "Memory", "Tool"] # Define items that need terminal colors

color_mapping = get_color_mapping(items) # Assign supported colors to the items

print(color_mapping) # Display the generated item-to-color mapping

# Exclude selected colors:
filtered_mapping = get_color_mapping( # Generate a mapping with excluded colors
    items=items, # Provide the items to color
    excluded_colors=["red", "pink"], # Exclude red and pink from the available colors
) # Finish generating the mapping

print(filtered_mapping) # Display the filtered color mapping
````

````python
# 2. Create colored text
colored_text = get_colored_text( # Create an ANSI-formatted string
    text="The retriever returned three documents.", # Provide the text to format
    color="green", # Apply the supported green color
) # Finish formatting the text

print(colored_text) # Display the colored, bold, and italic text
````

````python
# 3. Create bold text

bold_text = get_bolded_text( # Create an ANSI-formatted bold string
    text="LangChain execution report", # Provide the text to make bold
) # Finish formatting the text

print(bold_text) # Display the bold text
````

````python
# 4. Print text with different colors
print_text("Starting retrieval...\n", color="blue") # Print a blue status message
print_text("Documents retrieved successfully.\n", color="green") # Print a green success message
print_text("Warning: one document was empty.\n", color="yellow") # Print a yellow warning message
print_text("Request failed.\n", color="red") # Print a red failure message
````

````python
# 5. Use the generated mapping to print component
for component, color in color_mapping.items(): # Iterate through each component and assigned color
    print_text( # Print the component using its assigned color
        text=f"{component} completed\n", # Create the component status message
        color=color, # Apply the assigned color
    ) # Finish printing the component status
````

````python
# 6. Write formatted output to a file
with open("langchain_status.txt", "w", encoding="utf-8") as file: # Open a text file for writing
    print_text( # Write text through the LangChain utility
        text="LangChain workflow completed", # Provide the text to write
        end="\n", # Add a newline after the text
        file=file, # Send the text to the opened file
    ) # Finish writing the text

print("Status saved to langchain_status.txt") # Confirm that the file was written
````
