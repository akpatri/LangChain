Developer-facing statements defined in `libs/core/langchain_core/utils/formatting.py`.

# `StrictFormatter: Formatter`

String formatter that requires replacement values to be supplied through keyword arguments.

The inherited `Formatter.format` method uses the overridden `vformat`, so positional arguments are rejected there as well.

## Methods

### `vformat`

Formats a string using keyword arguments only.

```python
vformat(
    self,
    format_string: str, # String containing replacement fields
    args: Sequence[Any], # Positional arguments; must be empty
    kwargs: Mapping[str, Any], # Keyword values used for substitution
) -> str # Formatted string
```

Raises `ValueError` when `args` contains one or more positional arguments. Otherwise, formatting is delegated to `Formatter.vformat`.

### `validate_input_variables`

Validates that the supplied variable names can satisfy the named placeholders in a format string.

```python
validate_input_variables(
    self,
    format_string: str, # String containing replacement fields to validate
    input_variables: list[str], # Variable names available for substitution
) -> None
```

Each supplied variable name is assigned a dummy string and the format string is processed through the inherited formatter.

Raises `KeyError` when a named placeholder is missing from `input_variables`.

````python
from langchain_core.utils.formatting import StrictFormatter # Import the strict string formatter


formatter = StrictFormatter() # Create a StrictFormatter object

template = "Hello, {name}! Your course is {course}." # Define a template with named placeholders

result = formatter.format( # Format the template using keyword arguments
    template, # Provide the template string
    name="Saad", # Supply the value for the name placeholder
    course="LangChain", # Supply the value for the course placeholder
) # Finish formatting the string

print(result) # Display the formatted string
````

````python
# Validate template variables
template = "Hello, {name}! Your role is {role}." # Define a template containing two placeholders

formatter.validate_input_variables( # Validate the available variable names
    format_string=template, # Provide the template to validate
    input_variables=["name", "role"], # Provide all required placeholder names
) # Finish validating the template

print("Template variables are valid.") # Confirm successful validation
````

````python
# Missing variable example
template = "Hello, {name}! Your role is {role}." # Define a template containing two placeholders

try: # Begin exception handling
    formatter.validate_input_variables( # Validate an incomplete variable list
        format_string=template, # Provide the template to validate
        input_variables=["name"], # Omit the required role variable
    ) # Finish the validation call

except KeyError as error: # Catch the missing-placeholder error
    print("Missing variable:", error) # Display the missing variable
````

````python
# Positional arguments are rejected
try: # Begin exception handling
    result = formatter.format( # Attempt to format using positional arguments
        "Hello, {}!", # Define a positional placeholder
        "Saad", # Pass a positional value
    ) # Finish the formatting call

except ValueError as error: # Catch the positional-argument error
    print(error) # Display the validation message
````
