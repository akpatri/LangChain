# String Utilities API Reference

Developer-facing statements defined in `libs/core/langchain_core/utils/strings.py`.

# `stringify_value`

Converts a value into a string representation.

```python
stringify_value(
    val: Any, # Value to stringify
) -> str # String representation
```

Strings are returned unchanged. Dictionaries are prefixed with a newline and formatted through `stringify_dict`. Lists are converted item by item through `stringify_value` and joined with newline characters. Other values are converted with `str`.

---

# `stringify_dict`

Converts a dictionary into newline-terminated `key: value` entries.

```python
stringify_dict(
    data: dict[Any, Any], # Dictionary to stringify
) -> str # Concatenated key-value representation
```

Values are formatted through `stringify_value`, and dictionary iteration order is preserved. Each entry ends with a newline character.

---

# `comma_list`

Converts an iterable into a comma-and-space-separated string.

```python
comma_list(
    items: Iterable[Any], # Iterable whose items should be converted
) -> str # Comma-separated representation
```

Each item is converted with `str`. An empty iterable returns an empty string.

---

# `sanitize_for_postgres`

Removes or replaces NUL bytes that PostgreSQL text fields cannot store.

```python
sanitize_for_postgres(
    text: str, # Text to sanitize
    replacement: str = "", # String used to replace each NUL byte
) -> str # Sanitized text
```

The function replaces every `\x00` character using `str.replace`. By default, NUL bytes are removed.

````python
# 1. Convert different values using stringify_value()
from langchain_core.utils.strings import stringify_value # Import the general value-stringification function


text_value = stringify_value("LangChain") # Convert a string without changing it
number_value = stringify_value(42) # Convert an integer using str
list_value = stringify_value(["Python", "SQL", "LangChain"]) # Convert list items and join them with newlines
dictionary_value = stringify_value({"name": "Saad", "role": "Developer"}) # Convert a dictionary into key-value lines

print("String value:") # Display a heading
print(text_value) # Display the unchanged string

print("\nNumber value:") # Display a heading
print(number_value) # Display the converted integer

print("\nList value:") # Display a heading
print(list_value) # Display each list item on a separate line

print("\nDictionary value:") # Display a heading
print(dictionary_value) # Display the formatted dictionary
````

````python

# 2. Convert a dictionary using stringify_dict()
from langchain_core.utils.strings import stringify_dict # Import the dictionary-stringification function

employee = { # Create a sample employee dictionary
    "name": "Saad", # Store the employee name
    "skills": ["Python", "SQL", "LangChain"], # Store a list of skills
    "experience": 1, # Store the experience value
} # Finish creating the dictionary

formatted_employee = stringify_dict(employee) # Convert the dictionary into key-value text

print(formatted_employee) # Display the formatted dictionary
# The output follows the dictionary’s insertion order and ends each entry with a newline.
````

````python
# 3. Create a comma-separated list using comma_list()
from langchain_core.utils.strings import comma_list # Import the comma-list formatting function

technologies = ["Python", "FastAPI", "MongoDB", "LangChain"] # Create a list of technologies
formatted_technologies = comma_list(technologies) # Join the values with a comma and space
print(formatted_technologies) # Display the comma-separated result

# It also works with non-string values:
numbers = comma_list([10, 20, 30, 40]) # Convert integers and join them with commas

print(numbers) # Display the formatted numbers
````

````python
# 4. Remove PostgreSQL-incompatible NUL bytes
from langchain_core.utils.strings import sanitize_for_postgres # Import the PostgreSQL sanitization function


raw_text = "LangChain\x00Python\x00PostgreSQL" # Create text containing NUL-byte characters

clean_text = sanitize_for_postgres(raw_text) # Remove all NUL bytes using the default replacement

print(repr(raw_text)) # Display the original text including escape characters
print(repr(clean_text)) # Display the sanitized text
````

````python
# 5. Replace NUL bytes instead of removing them
from langchain_core.utils.strings import sanitize_for_postgres # Import the sanitization function


raw_text = "first\x00second\x00third" # Create text containing multiple NUL bytes

clean_text = sanitize_for_postgres( # Replace the NUL bytes with visible separators
    text=raw_text, # Provide the text to sanitize
    replacement=" | ", # Replace every NUL byte with this string
) # Finish sanitizing the text

print(clean_text) # Display the sanitized result
````
