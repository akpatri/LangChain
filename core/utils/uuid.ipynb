# `uuid7`

Generates a UUIDv7 from a Unix timestamp in nanoseconds and random bits.

```python
uuid7(
    nanoseconds: int | None = None, # Unix timestamp in nanoseconds, or None to use the current time
) -> UUID # Generated UUIDv7 object
```

UUIDs generated within the same millisecond are monotonic.

When `nanoseconds` is `None`, the function delegates directly to `uuid_utils.compat.uuid7()`. When a timestamp is provided, it is split into whole seconds and remaining nanoseconds and passed as `timestamp` and `nanos` to that implementation.

````python
# 1. Generate a UUIDv7 using the current time
from langchain_core.utils.uuid import uuid7 # Import LangChain's UUIDv7 generator


generated_id = uuid7() # Generate a UUIDv7 using the current system time

print("UUID:", generated_id) # Display the generated UUID
print("Python type:", type(generated_id).__name__) # Display the returned object type
print("UUID version:", generated_id.version) # Confirm that the UUID version is 7
````

````python

# 2. Generate a UUIDv7 from a nanosecond timestamp
import time # Import utilities for working with timestamps

from langchain_core.utils.uuid import uuid7 # Import the UUIDv7 generator


timestamp_ns = time.time_ns() # Get the current Unix timestamp in nanoseconds

generated_id = uuid7( # Generate a UUID using the supplied timestamp
    nanoseconds=timestamp_ns, # Provide the Unix timestamp in nanoseconds
) # Finish generating the UUID

print("Timestamp:", timestamp_ns) # Display the supplied timestamp
print("UUID:", generated_id) # Display the generated UUID
print("UUID version:", generated_id.version) # Confirm that the generated UUID is version 7
````

````python
# 3. Generate multiple monotonic UUIDs in the same millisecond
from langchain_core.utils.uuid import uuid7 # Import the UUIDv7 generator


same_timestamp_ns = 1_750_000_000_123_000_000 # Create one fixed nanosecond timestamp

generated_ids = [ # Generate several UUIDs using the same timestamp
    uuid7(nanoseconds=same_timestamp_ns) # Generate one monotonic UUIDv7
    for _ in range(5) # Repeat the generation five times
] # Finish creating the UUID list

for generated_id in generated_ids: # Iterate through the generated UUIDs
    print(generated_id) # Display each UUID

print("Already sorted:", generated_ids == sorted(generated_ids)) # Check their monotonic ordering
````

````python
# 4. Use UUIDv7 values as record identifiers
from langchain_core.utils.uuid import uuid7 # Import the UUIDv7 generator


documents = [ # Create sample records with UUIDv7 identifiers
    {"id": str(uuid7()), "content": "Introduction to LangChain"}, # Create the first document
    {"id": str(uuid7()), "content": "Working with retrievers"}, # Create the second document
    {"id": str(uuid7()), "content": "Building agents"}, # Create the third document
] # Finish creating the document list

for document in documents: # Iterate through the records
    print(document["id"], "-", document["content"]) # Display each identifier and document content
````
