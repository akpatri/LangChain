# LangChain JSON Serialization Reference

Developer-facing statements defined in `langchain_core.load.dump`.

# `default`

Returns a JSON-serializable representation for an object, using LangChain serialization when the object is `Serializable` and a not-implemented serialization record otherwise.

```python
default(
    obj: Any, # Object to serialize
) -> Any # JSON-serializable object or serialized not-implemented representation
```

---

# `dumps`

Serializes an object to a JSON string.

```python
dumps(
    obj: Any, # Object to serialize
    *,
    pretty: bool = False, # Whether to indent the JSON output
    **kwargs: Any, # Additional arguments passed to json.dumps
) -> str # JSON string representation of the object
```

## Behaviour

Plain dictionaries containing an `"lc"` key are escaped so they cannot be confused with LangChain serialization records. When `pretty=True`, the output uses an indentation of two spaces unless `indent` is supplied through `kwargs`.

A parsed Pydantic model stored in an `AIMessage` inside a `ChatGeneration` is converted to a dictionary on a deep copy before serialization.

## Exceptions

Raises `ValueError` when `default` is supplied through `kwargs`.

---

# `dumpd`

Serializes an object to a JSON-compatible Python representation.

```python
dumpd(
    obj: Any, # Object to serialize
) -> Any # Representation that can be serialized with json.dumps
```

## Behaviour

Plain dictionaries containing an `"lc"` key are escaped so they cannot be confused with LangChain serialization records.

A parsed Pydantic model stored in an `AIMessage` inside a `ChatGeneration` is converted to a dictionary on a deep copy before serialization.

````python
from langchain_core.load.dump import dumpd, dumps # Import both dump functions

data = {"name": "Saad", "age": 22} # Create sample Python data

json_string = dumps(data) # Convert the data into a JSON string
python_data = dumpd(data) # Convert the data into JSON-compatible Python data

print("dumps output:", json_string) # Print the JSON string
print("dumpd output:", python_data) # Print the Python dictionary
````
