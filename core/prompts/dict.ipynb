# Dictionary Prompt Template

The `dict` module provides `DictPromptTemplate`, a serializable Runnable that represents a prompt as a nested dictionary. It detects variables inside string values, formats them using either f-string or Mustache syntax, and preserves the dictionary structure in the formatted output.

# DictPromptTemplate

`DictPromptTemplate` represents a prompt using a dictionary. It recursively recognizes and replaces variables in dictionary values, including strings contained inside nested dictionaries, lists, and tuples. Variables used as dictionary keys are not recognized.

It inherits from `RunnableSerializable[dict[str, Any], dict[str, Any]]`.

### Attributes

1. `template`: Stores the dictionary structure used as the prompt template.
   * **Type:**
     ```python
     dict[str, Any]
     ```

2. `template_format`: Specifies the formatting style used for variables inside string values.
   * **Type:**
     ```python
     Literal["f-string", "mustache"]
     ```

### Properties

1. `input_variables`: Returns the unique variable names found recursively inside the template's string values.
   * **Type:**
     ```python
     list[str]
     ```

### Methods

1. `validate_template`: Validates the template after model creation by checking the variables contained in its nested structure.
   * **Syntax:**
     ```python
     validate_template(
         self
     ) -> DictPromptTemplate
     ```

2. `format`: Formats the dictionary template using the supplied input values.
   * **Syntax:**
     ```python
     format(
         self,
         **kwargs: Any # Values used to replace template variables
     ) -> dict[str, Any]
     ```

3. `aformat`: Asynchronously formats the dictionary template using the supplied input values.
   * **Syntax:**
     ```python
     aformat(
         self,
         **kwargs: Any # Values used to replace template variables
     ) -> dict[str, Any]
     ```

4. `invoke`: Invokes the prompt synchronously with an input dictionary and returns the formatted dictionary.
   * **Syntax:**
     ```python
     invoke(
         self,
         input: dict[str, Any], # Input values used to format the prompt
         config: RunnableConfig | None = None, # Runtime configuration for the Runnable
         **kwargs: Any # Additional arguments
     ) -> dict[str, Any]
     ```

5. `is_lc_serializable`: Returns `True` to indicate that the class supports LangChain serialization.
   * **Syntax:**
     ```python
     is_lc_serializable(
         cls
     ) -> bool
     ```

6. `get_lc_namespace`: Returns the namespace used to identify the class during LangChain serialization.
   * **Syntax:**
     ```python
     get_lc_namespace(
         cls
     ) -> list[str]
     ```

7. `pretty_repr`: Returns a human-readable representation of the prompt template.

   This method is declared by the class but is not implemented and raises `NotImplementedError`.
   * **Syntax:**
     ```python

     ) -> str
     ```
