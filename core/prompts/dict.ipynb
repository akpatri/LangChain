# Dictionary Prompt Template

The `dict` module defines `DictPromptTemplate`, a serializable Runnable that represents a prompt as a nested dictionary. It detects variables inside string values, formats them using f-string or Mustache syntax, and preserves the original dictionary structure in the formatted output.

# DictPromptTemplate: `RunnableSerializable[dict[str, Any], dict[str, Any]]`

`DictPromptTemplate` represents a prompt using a dictionary. It recursively detects and replaces variables in string values contained inside nested dictionaries, lists, and tuples. Variables used as dictionary keys are not detected.

## Fields

1. `template`:`dict[str, Any]`:= Stores the nested dictionary structure used as the prompt template.

2. `template_format`:`Literal["f-string", "mustache"]`:= Specifies the formatting style used for variables inside string values.

## Properties

1. `input_variables`:`list[str]`:= Returns the unique variable names found recursively inside the template's string values.

## Validators

1. `validate_template`:= Validates the nested template structure and determines the variables contained in its string values.

   This validator runs automatically after model creation.

   **Syntax**

   ```python
   validate_template(
       self # Prompt template instance to validate
   ) -> DictPromptTemplate
   ```

## Methods

1. `format`:= Formats the nested dictionary template using the supplied input values while preserving its dictionary, list, and tuple structure.

   **Syntax**

   ```python
   format(
       self, # Prompt template instance
       **kwargs: Any # Values used to replace template variables
   ) -> dict[str, Any]
   ```

2. `aformat`:= Asynchronously formats the nested dictionary template using the supplied input values.

   **Syntax**

   ```python
   async aformat(
       self, # Prompt template instance
       **kwargs: Any # Values used to replace template variables
   ) -> dict[str, Any]
   ```

3. `invoke`:= Invokes the prompt synchronously with an input dictionary and returns the formatted dictionary.

   **Syntax**

   ```python
   invoke(
       self, # Prompt template instance
       input: dict[str, Any], # Input values used to format the prompt
       config: RunnableConfig | None = None, # Runtime configuration for the Runnable
       **kwargs: Any # Additional invocation arguments
   ) -> dict[str, Any]
   ```

4. `is_lc_serializable`:= Returns `True` to indicate that the class supports LangChain serialization.

   **Syntax**

   ```python
   @classmethod
   is_lc_serializable(
       cls # DictPromptTemplate class
   ) -> bool
   ```

5. `get_lc_namespace`:= Returns the namespace used to identify the class during LangChain serialization.

   **Syntax**

   ```python
   @classmethod
   get_lc_namespace(
       cls # DictPromptTemplate class
   ) -> list[str]
   ```

6. `pretty_repr`:= Returns a human-readable representation of the prompt template.

   This method is not implemented and raises `NotImplementedError`.

   **Syntax**

   ```python
   pretty_repr(
       self, # Prompt template instance
       *,
       html: bool = False # Whether to format the representation as HTML
   ) -> str
   ```


````python
from langchain_core.prompts import DictPromptTemplate

prompt = DictPromptTemplate(
    template={
        "type": "user_profile",
        "message": "Hello, {name}!",
        "details": {
            "city": "{city}",
            "skills": ["{primary_skill}", "Python"],
            "summary": ("{name} lives in {city}.",)
        },
        "{unchanged_key}": "Variables in keys are not formatted"
    },
    template_format="f-string"
)

# Display the variables detected inside string values
print(sorted(prompt.input_variables))
# ['city', 'name', 'primary_skill']

formatted_prompt = prompt.format(
    name="Saad",
    city="Delhi",
    primary_skill="SQL"
)

print(formatted_prompt)


invoked_prompt = prompt.invoke(
    {
        "name": "Arfin",
        "city": "Bengaluru",
        "primary_skill": "Python"
    }
)

print(invoked_prompt["message"])
# Hello, Arfin!
````
