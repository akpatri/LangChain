# Image Prompt Template

The `image.py` module defines an image prompt template for multimodal models. It formats image URLs and optional detail settings, supports multiple string-template formats, and returns image-specific prompt values.

## Classes

- `ImagePromptTemplate`: Represents a URL-based image prompt template for a multimodal model.

## `ImagePromptTemplate`

`ImagePromptTemplate` formats image prompt data using a dictionary template. It supports `f-string`, `mustache`, and `jinja2` formatting, requires a valid image URL, optionally includes an image-detail setting, and prevents loading images from local file paths.

### Bases

- `BasePromptTemplate[ImageURL]`

### Attributes

1. `template`: Stores the dictionary used to construct the image prompt.
   * **Type:**
     ```python
     dict[str, Any]
     ```
   * **Default:**
     ```python
     {}
     ```

2. `template_format`: Specifies the formatting style used for string values inside the template.
   * **Type:**
     ```python
     PromptTemplateFormat
     ```
   * **Default:**
     ```python
     "f-string"
     ```
   * **Supported values:** `"f-string"`, `"mustache"`, and `"jinja2"`.

### Properties

1. `_prompt_type`: Returns the serialization type key used for the image prompt.
   * **Type:**
     ```python
     str
     ```
   * **Value:**
     ```python
     "image-prompt"
     ```
