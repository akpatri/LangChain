# Image Prompt Template

The `image.py` module defines an image prompt template for multimodal models. It formats image URLs and optional detail settings using supported string-template formats and returns image-specific prompt values.

# ImagePromptTemplate: `BasePromptTemplate[ImageURL]`

`ImagePromptTemplate` represents a URL-based image prompt template for a multimodal model. It formats image data from a dictionary template, requires a valid image URL, optionally includes an image-detail setting, and prevents images from being loaded from local file paths.

**Syntax**

```python
ImagePromptTemplate(
    self, # Image prompt template instance
    **kwargs: Any # Image prompt template configuration
)
```

## Fields

1. `template`:`dict[str, Any]`:= Stores the dictionary used to construct the image prompt. Its default value is an empty dictionary.

2. `template_format`:`PromptTemplateFormat`:= Specifies the formatting style used for string values inside the template. Its default value is `"f-string"`. Supported values are `"f-string"`, `"mustache"`, and `"jinja2"`.

## Properties

1. `_prompt_type`:`str`:= Returns the serialization type key used for the image prompt. Its value is `"image-prompt"`.

## Methods

1. `__init__`:= Initialises the image prompt template and validates its input variables and template strings.

   The input variables cannot contain `url`, `path`, or `detail`.

   **Syntax**

   ```python
   __init__(
       self, # Image prompt template instance
       **kwargs: Any # Image prompt template configuration
   ) -> None
   ```

2. `get_lc_namespace`:= Returns the LangChain serialization namespace assigned to the image prompt template.

   **Syntax**

   ```python
   @classmethod
   get_lc_namespace(
       cls # ImagePromptTemplate class
   ) -> list[str]
   ```

3. `format_prompt`:= Synchronously formats the supplied values and returns an `ImagePromptValue`.

   **Syntax**

   ```python
   format_prompt(
       self, # Image prompt template instance
       **kwargs: Any # Values used to format the image prompt
   ) -> PromptValue
   ```

4. `aformat_prompt`:= Asynchronously formats the supplied values and returns an `ImagePromptValue`.

   **Syntax**

   ```python
   async aformat_prompt(
       self, # Image prompt template instance
       **kwargs: Any # Values used to format the image prompt
   ) -> PromptValue
   ```

5. `format`:= Synchronously formats the template and returns an image URL dictionary.

   An error is raised when the URL is missing, is not a string, or refers to a local file path.

   **Syntax**

   ```python
   format(
       self, # Image prompt template instance
       **kwargs: Any # Values used to format the URL and detail fields
   ) -> ImageURL
   ```

6. `aformat`:= Asynchronously formats the template and returns an image URL dictionary.

   **Syntax**

   ```python
   async aformat(
       self, # Image prompt template instance
       **kwargs: Any # Values used to format the URL and detail fields
   ) -> ImageURL
   ```

7. `pretty_repr`:= Declares a method for returning a human-readable representation of the image prompt template.

   This method is not implemented and raises `NotImplementedError`.

   **Syntax**

   ```python
   pretty_repr(
       self, # Image prompt template instance
       html: bool = False # Whether to return an HTML-formatted representation
   ) -> str
   ```


````python
from langchain_core.prompts import ImagePromptTemplate # Import the image prompt template class

image_prompt = ImagePromptTemplate( # Create an image prompt template
    input_variables=["image_name"], # Declare the variable used inside the image URL
    template={ # Define the image prompt dictionary
        "url": "https://example.com/images/{image_name}.jpg", # Create a variable-based image URL
        "detail": "high" # Request a high-detail image
    }, # End the template dictionary
    template_format="f-string" # Use f-string formatting syntax
) # Finish creating the image prompt template

print(image_prompt._prompt_type) # Display the serialization type as image-prompt
print(image_prompt.get_lc_namespace()) # Display the LangChain serialization namespace

formatted_image = image_prompt.format( # Format the template synchronously
    image_name="mountain" # Supply the image-name variable
) # Finish synchronous formatting
print(formatted_image) # Display the formatted image URL dictionary

prompt_value = image_prompt.format_prompt( # Create an ImagePromptValue synchronously
    image_name="forest" # Supply the image-name variable
) # Finish creating the prompt value
print(prompt_value) # Display the generated image prompt value

async_image = await image_prompt.aformat( # Format the template asynchronously in Jupyter
    image_name="ocean" # Supply the image-name variable
) # Finish asynchronous formatting
print(async_image) # Display the asynchronously formatted dictionary

async_prompt_value = await image_prompt.aformat_prompt( # Create an ImagePromptValue asynchronously
    image_name="desert" # Supply the image-name variable
) # Finish creating the asynchronous prompt value
print(async_prompt_value) # Display the asynchronous image prompt value

try: # Test the unsupported local-file-path behaviour
    local_prompt = ImagePromptTemplate( # Create a template containing a local path
        input_variables=[], # Declare that no formatting variables are required
        template={"url": "./images/photo.jpg"} # Supply a prohibited local image path
    ) # Finish creating the local-path template
    print(local_prompt.format()) # Attempt to format the local image path
except ValueError as error: # Catch the expected local-path error
    print(error) # Display the validation error

try: # Test the unimplemented representation method
    print(image_prompt.pretty_repr()) # Attempt to generate a readable representation
except NotImplementedError as error: # Catch the expected error
    print(type(error).__name__) # Display the error type
````
