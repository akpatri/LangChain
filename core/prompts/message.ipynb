# Message Prompt Templates

The `message.py` module defines the abstract base interface for message prompt templates. It provides common methods for serialization, synchronous and asynchronous message formatting, readable display, and combining message templates into chat prompt templates.

## Classes

- `BaseMessagePromptTemplate`: Abstract base class for prompt templates that format inputs into lists of chat messages.

## `BaseMessagePromptTemplate`

`BaseMessagePromptTemplate` defines the shared interface implemented by message prompt templates. Subclasses must specify their required input variables and implement synchronous message formatting.

### Bases

- `Serializable`
- `ABC`

### Properties

1. `input_variables`: Returns the names of variables required to format the message prompt template.

   Subclasses must implement this abstract property.
   * **Type:**
     ```python
     list[str]
     ```

### Methods

1. `is_lc_serializable`: Indicates that the message prompt template supports LangChain serialization.
   * **Syntax:**
     ```python
     is_lc_serializable(
         cls
     ) -> bool
     ```

2. `get_lc_namespace`: Returns the LangChain serialization namespace for message prompt templates.
   * **Syntax:**
     ```python
     get_lc_namespace(
         cls
     ) -> list[str]
     ```

3. `format_messages`: Synchronously formats the supplied keyword arguments into a list of messages.

   Subclasses must implement this abstract method.
   * **Syntax:**
     ```python
     format_messages(
         self,
         **kwargs: Any # Values used to format the messages
     ) -> list[BaseMessage]
     ```

4. `aformat_messages`: Asynchronously formats the supplied keyword arguments into a list of messages. The default implementation calls `format_messages`.
   * **Syntax:**
     ```python
     aformat_messages(
         self,
         **kwargs: Any # Values used to format the messages
     ) -> list[BaseMessage]
     ```

5. `pretty_repr`: Returns a human-readable representation of the message prompt template.

   The base implementation is not implemented and raises `NotImplementedError`.
   * **Syntax:**
     ```python
     pretty_repr(
         self,
         html: bool = False # Whether to return an HTML-formatted representation
     ) -> str
     ```

6. `pretty_print`: Prints a human-readable representation of the message prompt template. HTML formatting is selected automatically in interactive environments.
   * **Syntax:**
     ```python
     pretty_print(
         self
     ) -> None
     ```

7. `__add__`: Combines this message prompt template with another compatible prompt object and returns a `ChatPromptTemplate`.
   * **Syntax:**
     ```python
     __add__(
         self,
         other: Any # Prompt template or prompt-compatible object to combine
     ) -> ChatPromptTemplate
     ```
