# Message Prompt Templates

The `message.py` module defines the abstract base interface for message prompt templates. It provides shared methods for serialization, synchronous and asynchronous message formatting, readable display, and combining message templates into chat prompt templates.

# BaseMessagePromptTemplate: `Serializable`, `ABC`

`BaseMessagePromptTemplate` is an abstract base class for prompt templates that format supplied inputs into lists of chat messages. Subclasses must define their required input variables and implement synchronous message formatting.

This abstract class is not intended to be instantiated directly.

## Properties

1. `input_variables`:`list[str]`:= Returns the names of variables required to format the message prompt template.

   Subclasses must implement this abstract property.

## Methods

1. `is_lc_serializable`:= Returns `True` to indicate that message prompt templates support LangChain serialization.

   **Syntax**

   ```python
   @classmethod
   is_lc_serializable(
       cls # Message prompt template class
   ) -> bool
   ```

2. `get_lc_namespace`:= Returns the LangChain serialization namespace assigned to message prompt templates.

   **Syntax**

   ```python
   @classmethod
   get_lc_namespace(
       cls # Message prompt template class
   ) -> list[str]
   ```

3. `format_messages`:= Synchronously formats the supplied keyword arguments into a list of chat messages.

   Subclasses must implement this abstract method.

   **Syntax**

   ```python
   format_messages(
       self, # Message prompt template instance
       **kwargs: Any # Values used to format the messages
   ) -> list[BaseMessage]
   ```

4. `aformat_messages`:= Asynchronously formats the supplied keyword arguments into a list of chat messages.

   The default implementation calls `format_messages()`.

   **Syntax**

   ```python
   async aformat_messages(
       self, # Message prompt template instance
       **kwargs: Any # Values used to format the messages
   ) -> list[BaseMessage]
   ```

5. `pretty_repr`:= Returns a human-readable representation of the message prompt template.

   The base implementation is not implemented and raises `NotImplementedError`.

   **Syntax**

   ```python
   pretty_repr(
       self, # Message prompt template instance
       html: bool = False # Whether to return an HTML-formatted representation
   ) -> str
   ```

6. `pretty_print`:= Prints a human-readable representation of the message prompt template.

   HTML formatting is selected automatically when the method runs in an interactive environment.

   **Syntax**

   ```python
   pretty_print(
       self # Message prompt template instance
   ) -> None
   ```

7. `__add__`:= Combines the current message prompt template with another prompt-compatible object and returns a `ChatPromptTemplate`.

   **Syntax**

   ```python
   __add__(
       self, # Current message prompt template instance
       other: Any # Prompt template or compatible object to combine
   ) -> ChatPromptTemplate
   ```


````python
from langchain_core.prompts import HumanMessagePromptTemplate # Import a concrete message prompt template

prompt = HumanMessagePromptTemplate.from_template( # Create a human message prompt template
    "Hello, {name}!" # Define the message content with one variable
) # Finish creating the prompt template

print(prompt.input_variables) # Display the required input variables
print(prompt.is_lc_serializable()) # Check whether LangChain serialization is supported
print(prompt.get_lc_namespace()) # Display the LangChain serialization namespace

messages = prompt.format_messages(name="Saad") # Format the message synchronously
print(messages[0].content) # Display the formatted message content

async_messages = await prompt.aformat_messages(name="Arfin") # Format the message asynchronously in Jupyter
print(async_messages[0].content) # Display the asynchronously formatted content

print(prompt.pretty_repr()) # Display a readable representation of the template
prompt.pretty_print() # Print the readable representation automatically

chat_prompt = prompt + "How can I help you?" # Combine the message template with another prompt-compatible value
print(chat_prompt.format_messages(name="Saad")) # Format the combined ChatPromptTemplate
````
