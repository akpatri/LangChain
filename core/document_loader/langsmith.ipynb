Below is a complete Jupyter-friendly example of creating a custom BlobLoader. It reads all .txt files from a folder and lazily yields each file as a Blob.

# !pip install langchain-core#Install LangChain Core if it is not already installed

from collections.abc import Iterator#Import Iterator for the yield_blobs return type
from pathlib import Path#Import Path for working with files and folders
from langchain_core.document_loaders.blob_loaders import BlobLoader#Import the abstract BlobLoader class
from langchain_core.documents import Blob#Import Blob for representing raw file content


class TextFolderBlobLoader(BlobLoader):#Create a custom loader for text files inside a folder

    def __init__(self, folder_path: str) -> None:#Initialize the loader with a folder path
        self.folder_path = Path(folder_path)#Convert the supplied folder path into a Path object

    def yield_blobs(self) -> Iterator[Blob]:#Implement the required BlobLoader method
        for file_path in self.folder_path.glob("*.txt"):#Find every text file inside the folder
            blob = Blob.from_path(#Create a Blob from the current file
                path=file_path,#Provide the path of the text file
                mime_type="text/plain",#Specify that the file contains plain text
            )

            yield blob#Return the Blob lazily


sample_folder = Path("sample_text_files")#Create a Path object for the sample folder
sample_folder.mkdir(exist_ok=True)#Create the folder if it does not already exist

file_1 = sample_folder / "python.txt"#Create the path for the first sample file
file_1.write_text("Python is a high-level programming language.", encoding="utf-8")#Write content into the first file

file_2 = sample_folder / "langchain.txt"#Create the path for the second sample file
file_2.write_text("LangChain helps developers build LLM applications.", encoding="utf-8")#Write content into the second file

loader = TextFolderBlobLoader("sample_text_files")#Create an instance of the custom BlobLoader

for blob in loader.yield_blobs():#Load and process each Blob lazily
    print(f"Source: {blob.source}")#Display the source path of the Blob
    print(f"Content: {blob.as_string()}")#Convert the Blob into text and display it
    print("-" * 50)#Display a separator after each Blob

# `LangSmithLoader: BaseLoader`

Loads examples from a LangSmith dataset as `Document` objects.

The selected example input becomes the document page content, while the complete example is stored in the document metadata. This is useful for building few-shot example retrievers.

## Fields

```python
content_key: list[str] # Nested input key path used for document content
format_content: Callable[..., str] # Function that converts selected input content to text
dataset_id: uuid.UUID | str | None # Dataset ID filter
dataset_name: str | None # Dataset name filter
example_ids: Sequence[uuid.UUID | str] | None # Example ID filter
as_of: datetime.datetime | str | None # Dataset version tag or timestamp
splits: Sequence[str] | None # Dataset split filters
inline_s3_urls: bool # Whether LangSmith should inline S3 URLs
offset: int # Number of matching examples to skip
limit: int | None # Maximum number of examples to return
metadata: dict[str, Any] | None # Metadata filter
filter: str | None # Structured LangSmith filter expression
```

## Constructor

```python
LangSmithLoader(
    *,
    dataset_id: uuid.UUID | str | None = None, # Dataset ID to filter by
    dataset_name: str | None = None, # Dataset name to filter by
    example_ids: Sequence[uuid.UUID | str] | None = None, # Example IDs to filter by
    as_of: datetime.datetime | str | None = None, # Version tag or timestamp to query
    splits: Sequence[str] | None = None, # Dataset splits to include
    inline_s3_urls: bool = True, # Whether to inline S3 URLs
    offset: int = 0, # Number of matching examples to skip
    limit: int | None = None, # Maximum number of examples to return
    metadata: dict[str, Any] | None = None, # Metadata values to filter by
    filter: str | None = None, # Structured filter string
    content_key: str = "", # Dot-separated input key used as page content
    format_content: Callable[..., str] | None = None, # Content-to-string formatter
    client: LangSmithClient | None = None, # Existing LangSmith client
    **client_kwargs: Any, # Arguments used to construct a LangSmith client
) -> None
```

When `content_key` is empty, the complete example input is used as page content. Dot-separated values such as `"first.second"` select nested input keys.

When `format_content` is omitted, strings are preserved and other values are JSON-encoded when possible.

Raises `ValueError` when both `client` and `client_kwargs` are supplied.

## Methods

Lazily retrieves matching LangSmith examples and converts them into documents.

```python
lazy_load(
    self,
) -> Iterator[Document] # Documents created from matching dataset examples
```

Each document contains the formatted selected input as its page content and the complete example as metadata. UUID and datetime metadata fields for the dataset, example, source run, creation time, and modification time are converted to strings.

Raises `ValueError` when `content_key` cannot be resolved because a key is missing or an intermediate value is not a mapping.
