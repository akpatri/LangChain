## What are LLMs and Chat Models?

Large Language Models (LLMs) are AI systems trained on massive text corpora.

Examples:
- GPT models
- Claude models
- Gemini models

Capabilities:
- Text generation
- Summarization
- Translation
- Question answering
- Reasoning
- Tool usage

Analogy:
Think of an LLM as a highly knowledgeable assistant that has read billions of pages.

LangChain provides a common interface for working with different model providers.

## Initializing Models

LangChain allows models from different providers to be initialized through a common API.

Why?
Without LangChain, every provider has a different SDK and syntax.

Typical flow:
1. Install provider package
2. Configure API key
3. Create model object
4. Invoke model

Example:

from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-5.4")


## Model Parameters

Important parameters:

1. model
   - Model name

2. temperature
   - Controls randomness
   - Low = deterministic
   - High = creative

3. max_tokens
   - Controls response length

4. timeout
   - Maximum wait time

5. max_retries
   - Retry failed requests

Interview Question:
Why use low temperature for production?
Answer: More consistent outputs.


## Invocation Methods

Three primary methods:

1. invoke()
   Single request

2. stream()
   Incremental output

3. batch()
   Multiple requests together

Example:

response = model.invoke("Explain transformers")

Use invoke() when you need a complete answer.


## Streaming

Streaming returns tokens as they are generated.

Benefits:
- Faster perceived response
- Better user experience

Example:

for chunk in model.stream("Explain AI"):
    print(chunk)

Real-world analogy:
Watching a live cricket match rather than waiting for the final score.


## Batch Processing

Batching sends multiple prompts simultaneously.

Benefits:
- Faster processing
- Better throughput
- Lower overhead

Example:

responses = model.batch([
    "Question 1",
    "Question 2",
    "Question 3"
])

Useful for:
- Classification
- Data extraction
- Large-scale processing


## Tool Calling

Tool calling allows the model to interact with external systems.

Examples:
- Weather API
- Database
- Calculator
- Search Engine

Flow:
User -> Model -> Tool -> Result -> Model -> User

Why important?
Models can access real-world data and perform actions.

Example tool:

@tool
def get_weather(city):
    return "Sunny"


## Structured Output

Structured output forces the model to produce data in a specific format.

Methods:
1. Pydantic
2. TypedDict
3. JSON Schema

Benefits:
- Reliable parsing
- Easier automation
- Better integration

Example:

class Movie(BaseModel):
    title: str
    year: int

model.with_structured_output(Movie)


## Model Profiles

Model profiles describe capabilities.

Examples:
- Context length
- Tool calling support
- Image support
- Reasoning support

Why useful?
Applications can dynamically choose models based on capabilities.


## Multimodal Models

Multimodal models handle more than text.

Input types:
- Text

- Video

Examples:
- GPT multimodal models

Use cases:
- Image analysis
- OCR

## Reasoning Models

Reasoning models perform multi-step thinking.

Use cases:
- Mathematics
- Coding
- Planning
- Logic

Benefits:
- Better accuracy
- Improved problem solving

Tradeoff:
Higher latency and cost.


## Prompt Caching

Prompt caching reduces cost and latency.

How?
Previously processed prompt portions can be reused.

Benefits:
- Faster responses
- Lower token cost

Useful for:
- Long system prompts
- Repeated workflows


## Rate Limiting

Providers limit request frequency.

LangChain offers rate limiters.

Benefits:
- Prevent API abuse
- Avoid provider errors
- Stable production systems

Example:

InMemoryRateLimiter(
    requests_per_second=1
)


## Token Usage

Tokens measure model consumption.

Metrics:
- Input tokens
- Output tokens
- Total tokens

Why important?
Token usage directly affects cost.

Monitor token usage in production applications.


## Dynamic Model Selection

Different tasks require different models.

Example strategy:

Simple query -> Cheap model
Complex query -> Powerful model

Benefits:
- Cost optimization
- Better performance
- Improved scalability
