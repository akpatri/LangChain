# System Information API Reference

Developer-facing statements defined in `libs/core/langchain_core/sys_info.py`.

# `print_sys_info`

Prints system, LangChain-package, optional-package, and related dependency information for debugging.

```python
print_sys_info(
    *,
    additional_pkgs: Sequence[str] = (), # Additional package names to inspect and include in the output
) -> None
```

The function writes its report to standard output. It includes the operating system, OS version, and Python version.

Package discovery includes installed modules whose names begin with `langchain` or `langgraph`, the known related packages `langsmith`, `deepagents`, and `deepagents-cli`, and any names supplied through `additional_pkgs`. Duplicate names are removed.

When present, `langchain_core`, `langchain`, `langchain_community`, and `langsmith` are displayed before the remaining packages. Packages that cannot be located are listed under `Optional packages not installed`. Installed packages whose versions cannot be determined are omitted from the version listing.

The function also prints installed direct dependencies declared by the selected packages, excluding package names already in the main package list. These dependency names are deduplicated and ordered case-insensitively.

````python
from langchain_core.sys_info import print_sys_info # Import the system-information utility


print_sys_info() # Print operating-system, Python, LangChain, and dependency information
````

````python
from langchain_core.sys_info import print_sys_info # Import the system-information utility


additional_packages = [ # Define extra packages to inspect
    "langchain-openai", # Check whether the LangChain OpenAI integration is installed
    "langchain-chroma", # Check whether the LangChain Chroma integration is installed
    "chromadb", # Check whether the Chroma database package is installed
] # Finish the additional package list

print_sys_info( # Print the complete environment report
    additional_pkgs=additional_packages, # Include the additional packages in the inspection
) # Finish the system-information call
````
