# Document Transformer Interface Reference

Developer-facing statements defined in `langchain_core.documents.transformers`.

# `BaseDocumentTransformer: ABC`

Abstract interface for transforming a sequence of `Document` objects into another sequence of documents.

A concrete subclass must implement `transform_documents`.

## Required subclass hooks

### `transform_documents`

Transforms a sequence of documents.

```python
transform_documents(
    self,
    documents: Sequence[Document], # Documents to transform
    **kwargs: Any, # Additional transformer-specific arguments
) -> Sequence[Document] # Transformed documents
```

## Methods

### `atransform_documents`

Asynchronously transforms a sequence of documents.

```python
async atransform_documents(
    self,
    documents: Sequence[Document], # Documents to transform
    **kwargs: Any, # Additional transformer-specific arguments
) -> Sequence[Document] # Transformed documents
```

The default implementation calls `transform_documents` through `run_in_executor` with no explicit executor. Subclasses may override this method to provide native asynchronous behaviour.

````python
#%pip install -U langchain-core#Install LangChain Core if it is not already installed

from collections.abc import Sequence#Import Sequence for type annotations
from typing import Any#Import Any for additional keyword arguments
from langchain_core.documents import Document#Import the LangChain Document class
from langchain_core.documents.transformers import BaseDocumentTransformer#Import the abstract transformer class


class TextCleaningTransformer(BaseDocumentTransformer):#Create a custom document transformer

    def transform_documents(self, documents: Sequence[Document], **kwargs: Any) -> Sequence[Document]:#Implement the required transformation method
        transformed_documents = []#Create an empty list for transformed documents

        for document in documents:#Process each input document
            cleaned_text = " ".join(document.page_content.split())#Remove repeated spaces and newline characters
            cleaned_text = cleaned_text.lower()#Convert the cleaned text to lowercase
            word_count = len(cleaned_text.split())#Count the words in the cleaned text
            updated_metadata = {**document.metadata, "word_count": word_count, "transformed": True}#Create updated metadata
            transformed_document = Document(page_content=cleaned_text, metadata=updated_metadata)#Create a new transformed Document
            transformed_documents.append(transformed_document)#Add the transformed Document to the result list

        return transformed_documents#Return all transformed documents


documents = [#Create sample input documents
    Document(page_content="  Python   is EASY to learn.  ", metadata={"source": "python.txt"}),#Create the first sample document
    Document(page_content="LangChain\nhelps build   LLM applications.", metadata={"source": "langchain.txt"}),#Create the second sample document
]#Finish the input document list

transformer = TextCleaningTransformer()#Create the custom transformer object
transformed_documents = transformer.transform_documents(documents)#Transform the sample documents

print(f"Total transformed documents: {len(transformed_documents)}")#Display the number of transformed documents
print("=" * 60)#Display a separator line

for document_number, document in enumerate(transformed_documents, start=1):#Process every transformed document
    print(f"Document {document_number}")#Display the document number
    print(f"Content: {document.page_content}")#Display the transformed content
    print(f"Metadata: {document.metadata}")#Display the updated metadata
    print("-" * 60)#Display a separator after each document
````
