# Generation Outputs

The `generation.py` module defines output objects for traditional text-based language models that accept strings and return generated text.

A generation may include the generated text together with optional provider-specific metadata such as finish reasons or token log probabilities.

# Generation

`Generation` represents a single text output produced by a traditional string-in, string-out language model.

It is used internally and may later be included in an `LLMResult` or converted into an `AIMessage`.

## Bases

- `Serializable`

## Attributes

1. `text`: Stores the text generated by the language model.
   * **Type:**
     ```python
     text: str
     ```

2. `generation_info`: Stores optional provider-specific response information.

   It may contain values such as the finish reason or token log probabilities.

   * **Type:**
     ```python
     generation_info: dict[str, Any] | None = None
     ```

3. `type`: Stores the generation type used during serialization and deserialization.
   * **Type:**
     ```python
     type: Literal["Generation"] = "Generation"
     ```

### Methods

1. `is_lc_serializable`: Indicates that `Generation` supports LangChain serialization.
   * **Syntax:**
     ```python
     @classmethod
     is_lc_serializable(
         cls
     ) -> bool
     ```

2. `get_lc_namespace`: Returns the LangChain serialization namespace for generation outputs.
   * **Syntax:**
     ```python
     @classmethod
     get_lc_namespace(
         cls
     ) -> list[str]
     ```

# GenerationChunk

`GenerationChunk` represents a partial text generation produced during streaming.

Multiple chunks can be concatenated to create a larger generation while combining their generated text and provider-specific metadata.

## Bases

- `Generation`

### Methods

1. `__add__`: Concatenates the current generation chunk with another generation chunk.

   The generated text values are joined, and the `generation_info` dictionaries are merged. A `TypeError` is raised when the supplied value is not a `GenerationChunk`.

   * **Syntax:**
     ```python
     __add__(
         self,
         other: GenerationChunk # Generation chunk to concatenate
     ) -> GenerationChunk
     ```
