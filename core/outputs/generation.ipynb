# `Generation: Serializable`

Represents a single text-generation output from a string-in, string-out language model.

This model is used internally and may later be mapped to an `LLMResult` and projected into an `AIMessage`. Chat-model users normally access output through `AIMessage` or `LLMResult`.

## Fields

```python
text: str # Generated text output
generation_info: dict[str, Any] | None = None # Raw provider response metadata, such as finish reasons or token log probabilities
type: Literal["Generation"] = "Generation" # Discriminator used exclusively for serialization
```

## Constructor

```python
Generation(
    *,
    text: str, # Generated text output
    generation_info: dict[str, Any] | None = None, # Raw provider response metadata
    type: Literal["Generation"] = "Generation", # Serialization discriminator
) -> None
```

## Methods

### `is_lc_serializable`

Reports that the class supports LangChain serialization.

```python
@classmethod
is_lc_serializable(
    cls,
) -> bool # Always True
```

### `get_lc_namespace`

Returns the serialization namespace for the class.

```python
@classmethod
get_lc_namespace(
    cls,
) -> list[str] # `["langchain", "schema", "output"]`
```

---



````python
from langchain_core.outputs import Generation # Import the Generation class

generation = Generation( # Create a text-generation result
    text="Python is easy to learn.", # Store the generated text
    generation_info={"finish_reason": "stop", "tokens": 6}, # Store model metadata
) # Finish creating the Generation object

print("Generated text:", generation.text) # Display the generated text
print("Generation info:", generation.generation_info) # Display the metadata
print("Type:", generation.type) # Display the serialization type
print("Serializable:", generation.is_lc_serializable()) # Check serialization support
print("Namespace:", generation.get_lc_namespace()) # Display the serialization namespace
````

# `GenerationChunk: Generation`

Represents a text-generation chunk that can be concatenated with another chunk.

It inherits `text`, `generation_info`, and `type` from `Generation`.

## Methods

### `__add__`

Concatenates this chunk with another `GenerationChunk`.

```python
__add__(
    self,
    other: GenerationChunk, # Chunk to concatenate
) -> GenerationChunk # Newly concatenated chunk
```

The returned chunk contains the two text values joined in order. Non-empty `generation_info` dictionaries are merged, and the result uses `None` when the merged metadata is empty.

Raises `TypeError` when `other` is not a `GenerationChunk`.

````python
from langchain_core.outputs import GenerationChunk # Import the GenerationChunk class

chunk1 = GenerationChunk( # Create the first text chunk
    text="Hello ", # Store the first part of the generated text
    generation_info={"part": 1}, # Store metadata for the first chunk
) # Finish creating the first chunk

chunk2 = GenerationChunk( # Create the second text chunk
    text="Saad!", # Store the second part of the generated text
    generation_info={"status": "finished"}, # Store metadata for the second chunk
) # Finish creating the second chunk

combined_chunk = chunk1 + chunk2 # Concatenate both chunks

print("Combined text:", combined_chunk.text) # Display the joined text
print("Combined metadata:", combined_chunk.generation_info) # Display the merged metadata
````
