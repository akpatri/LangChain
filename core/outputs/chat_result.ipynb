# LangChain Chat Result Reference

Developer-facing statements defined in `langchain_core.outputs.chat_result`.

# `ChatResult: BaseModel`

Represents the result of a chat-model call for a single prompt.

This container is used internally by some chat-model implementations before being mapped to an `LLMResult` and projected into an `AIMessage`. Users normally access chat-model information through `AIMessage` or `LLMResult`.

## Fields

```python
generations: list[ChatGeneration] # Candidate chat generations for the input prompt
llm_output: dict[str, Any] | None = None # Free-form model provider-specific output
```

`generations` is a list so a single prompt can have multiple candidate generations.

`llm_output` is not standardized, and its keys can vary by provider and over time. Prefer standardized information available on `AIMessage` when possible.

## Constructor

```python
ChatResult(
    *,
    generations: list[ChatGeneration], # Candidate chat generations for the input prompt
    llm_output: dict[str, Any] | None = None, # Free-form model provider-specific output
) -> None
```

````python
from langchain_core.messages import AIMessage # Import the AI message class
from langchain_core.outputs import ChatGeneration, ChatResult # Import the output classes

generation1 = ChatGeneration( # Create the first possible response
    message=AIMessage(content="Python is easy to learn.") # Store the first response
) # Finish the first generation

generation2 = ChatGeneration( # Create the second possible response
    message=AIMessage(content="Python is widely used in AI.") # Store the second response
) # Finish the second generation

result = ChatResult( # Create the result for one prompt
    generations=[generation1, generation2], # Store multiple candidate responses
    llm_output={"model": "demo-model", "tokens": 15}, # Store provider-specific information
) # Finish creating the chat result

print("First response:", result.generations[0].text) # Display the first response

for number, generation in enumerate(result.generations, start=1): # Visit every response
    print(f"Candidate {number}: {generation.text}") # Display each response

print("Model information:", result.llm_output) # Display provider-specific output
````
