# Chat Results

The `chat_result.py` module defines the internal result container used by chat-model implementations for a single input prompt.

A `ChatResult` may contain multiple candidate chat generations together with optional provider-specific metadata. It is generally converted into an `LLMResult` and then exposed to users through an `AIMessage`.

# ChatResult

`ChatResult` represents the result of one chat-model call for a single prompt.

LangChain users normally access chat-model output through `AIMessage` when using Runnable interfaces or through `LLMResult` in callbacks.

## Bases

- `BaseModel`

## Attributes

1. `generations`: Stores the candidate chat generations produced for the input prompt.

   The value is a list because a model may return multiple candidate responses for one prompt.

   * **Type:**
     ```python
     generations: list[ChatGeneration]
     ```

2. `llm_output`: Stores arbitrary provider-specific model output.

   This field is not standardized, and its keys may differ between providers or change over time. Standardized fields available through `AIMessage` should generally be preferred.

   * **Type:**
     ```python
     llm_output: dict[str, Any] | None = None
     ```
