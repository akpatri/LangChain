# System Messages

The `system.py` module defines messages used to provide instructions that guide a chat model's behaviour. A system message is usually placed at the beginning of a sequence of input messages.

# SystemMessage: `BaseMessage`

`SystemMessage` represents instructions used to establish or prime the behaviour of a chat model.

**Syntax**

```python
SystemMessage(
    self,
    content: str | list[str | dict[Any, Any]] | None = None, # Raw system-message content
    content_blocks: list[types.ContentBlock] | None = None, # Standardized content blocks
    **kwargs: Any # Additional message fields
)
```

## Fields
1. `type`:`Literal["system"]`:= Stores the message type used during serialization and deserialization. Its default value is `"system"`.

````python
from langchain_core.language_models.fake_chat_models import FakeListChatModel
from langchain_core.messages import HumanMessage, SystemMessage


system_message = SystemMessage(
    content="Answer every question in one short sentence.", # System instruction
    name="instruction", # Inherited BaseMessage field
    id="system-1" # Optional message identifier
)


print("System content:", system_message.content) # Display system instruction
print("System type:", system_message.type) # Displays "system"
print("System name:", system_message.name) # Display inherited name
print("System ID:", system_message.id) # Display message identifier
````


# SystemMessageChunk: `SystemMessage`, `BaseMessageChunk`

`SystemMessageChunk` represents a partial system message produced during streaming. It inherits message-chunk combination behaviour from `BaseMessageChunk`.

**Syntax**

```python
SystemMessageChunk(
    self,
    content: str | list[str | dict[Any, Any]] | None = None, # Raw partial system-message content
    content_blocks: list[types.ContentBlock] | None = None, # Standardized content blocks
    **kwargs: Any # Additional message fields
)
```

## Fields

1. `type`:`Literal["SystemMessageChunk"]`:= Stores the chunk-specific message type used during serialization and deserialization. Its default value is `"SystemMessageChunk"`.


````python
from langchain_core.messages import SystemMessageChunk


chunk1 = SystemMessageChunk(
    content="You are a helpful ", # First partial system message
    id="system-chunk-1" # Optional inherited message ID
)

chunk2 = SystemMessageChunk(
    content="Python assistant." # Second partial system message
)


combined_chunk = chunk1 + chunk2 # Combine both message chunks


print("Content:", combined_chunk.content) # Merged content
print("Type:", combined_chunk.type) # Displays "SystemMessageChunk"
print("ID:", combined_chunk.id) # Preserved ID from the first chunk
````
