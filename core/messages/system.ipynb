# System Messages

The `system.py` module defines messages used to provide instructions that guide a chat model's behaviour. A system message is usually placed at the beginning of a sequence of input messages.

# SystemMessage

`SystemMessage` represents instructions used to establish or prime the behaviour of a chat model.

## Bases

- `BaseMessage`

## Attributes

1. `type`: Stores the message type used during serialization and deserialization.
   * **Type:**
     ```python
     type: Literal["system"] = "system"
     ```

### Methods

1. `__init__`: Initializes a system message using raw content or standardized content blocks.

   When `content_blocks` is provided, it is stored as the message content. Otherwise, the value supplied through `content` is used.

   * **Syntax:**
     ```python
     __init__(
         self,
         content: str | list[str | dict[Any, Any]] | None = None, # Raw system-message content
         content_blocks: list[types.ContentBlock] | None = None, # Standardized content blocks
         **kwargs: Any # Additional message fields
     ) -> None
     ```

# SystemMessageChunk

`SystemMessageChunk` represents a partial system message produced during streaming. It inherits message-chunk combination behaviour from `BaseMessageChunk`.

## Bases

- `SystemMessage`
- `BaseMessageChunk`

## Attributes

1. `type`: Stores the chunk-specific message type used during serialization and deserialization.
   * **Type:**
     ```python
     type: Literal["SystemMessageChunk"] = "SystemMessageChunk"
     ```
