# Message Modifiers
The `modifier.py` module defines messages that modify an existing message history instead of adding normal conversational content.

# RemoveMessage: `BaseMessage`
`RemoveMessage` represents an instruction to remove another message from the conversation history using that message's identifier.
It does not contain normal message content. Its `content` field is automatically set to an empty string.
**Syntax**
```python
RemoveMessage(
    self,
    id: str, # ID of the message to remove
    **kwargs: Any # Additional message fields
)
```
## Fields
1. `type`:`Literal["remove"]`:= Stores the message type used during serialization and deserialization. Its default value is `"remove"`.
2. `id`:`str`:= Stores the identifier of the message that should be removed.
3. `content`:`str`:= Stores an empty string because `RemoveMessage` does not carry normal conversational content.


````python
from langchain_core.language_models.fake_chat_models import FakeListChatModel
from langchain_core.messages import HumanMessage, RemoveMessage
from langgraph.graph.message import add_messages

messages = [
    HumanMessage(
        content="My password is 1234.",
        id="message-1" # Message that will be removed
    ),
    HumanMessage(
        content="Explain Python lists.",
        id="message-2" # Message that will remain
    )
]


print("Before removal:")
for message in messages:
    print(message.id, ":", message.content)


remove_instruction = RemoveMessage(
    id="message-1" # ID of the message to remove
)

messages = add_messages(
    messages, # Existing message history
    remove_instruction # Removal instruction
)


print("\nAfter removal:")
for message in messages:
    print(message.id, ":", message.content)

````
