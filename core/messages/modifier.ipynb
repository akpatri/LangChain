# Message Modifiers

The `modifier.py` module defines messages that modify an existing message history rather than adding normal conversational content.

# RemoveMessage

`RemoveMessage` represents an instruction to remove another message from the conversation history using that message's identifier.

It does not carry normal message content. Its content is automatically stored as an empty string.

## Bases

- `BaseMessage`

## Attributes

1. `type`: Stores the message type used during serialization and deserialization.
   * **Type:**
     ```python
     type: Literal["remove"] = "remove"
     ```

2. `id`: Stores the identifier of the message that should be removed.
   * **Type:**
     ```python
     id: str
     ```

### Methods

1. `__init__`: Creates a removal message for the message with the specified identifier.

   The message content is automatically set to an empty string. A `ValueError` is raised when a non-empty `content` field is supplied through `kwargs`.

   * **Syntax:**
     ```python
     __init__(
         self,
         id: str, # ID of the message to remove
         **kwargs: Any # Additional message fields
     ) -> None
     ```
