# Human Messages
The `human.py` module defines messages that represent input sent by a user to a chat model. It also provides a streaming chunk variant for partial human messages.
# HumanMessage: `BaseMessage`
`HumanMessage` represents a message passed from a user to a chat model.

**Syntax**
  ```python
  HumanMessage(
      self,
      content: str | list[str | dict[Any, Any]] | None = None, # Raw message content
      content_blocks: list[types.ContentBlock] | None = None, # Standardized content blocks
      **kwargs: Any # Additional message fields
  )
  ```

## Fields

1. `type`:`Literal["human"]`:= Stores the message type used during serialization and deserialization. Its default value is `"human"`.

````python
from langchain_core.messages import HumanMessage


# Create a human message using raw text content
message1 = HumanMessage(
    content="Explain SQL joins.", # Raw message content
    name="student", # Optional message name
    id=101, # Numeric ID automatically converted to string
    additional_kwargs={
        "priority": "high" # Provider-specific data
    },
    response_metadata={
        "source": "user-input" # Message-related metadata
    }
)


print("Content:", message1.content) # Display raw message content
print("Type:", message1.type) # Display serialization type
print("Name:", message1.name) # Display optional message name
print("ID:", message1.id) # Display message identifier
print("ID type:", type(message1.id)) # Confirm ID is stored as string
print("Additional kwargs:", message1.additional_kwargs) # Display additional data
print("Response metadata:", message1.response_metadata) # Display metadata
print("Content blocks:", message1.content_blocks) # Display standardized content blocks
print("Text:", message1.text) # Display extracted textual content


# Create a human message using standardized content blocks
message2 = HumanMessage(
    content_blocks=[
        {
            "type": "text",
            "text": "Describe the given image."
        },
        {
            "type": "image",
            "url": "https://example.com/image.png"
        }
    ], # Standardized content blocks
    name="student" # Optional message name
)


print("\nContent-block message:")
print("Content:", message2.content) # Content created from content_blocks
print("Type:", message2.type) # Human message type
print("Text:", message2.text) # Extract only textual content
print("Content blocks:", message2.content_blocks) # Standardized content blocks


print("\nPretty representation:")
print(
    message1.pretty_repr(
        html=False # Use plain-text formatting
    )
)


print("\nPretty print:")
message1.pretty_print() # Print the formatted human message
````

# HumanMessageChunk: `HumanMessage`, `BaseMessageChunk`

`HumanMessageChunk` represents a partial human message produced during streaming. It inherits message-chunk combination behavior from `BaseMessageChunk`.

**Syntax**
```python
HumanMessageChunk(
    self,
    content: str | list[str | dict[Any, Any]] | None = None, # Raw partial message content
    content_blocks: list[types.ContentBlock] | None = None, # Standardized content blocks
    **kwargs: Any # Additional message fields
)
```

## Fields

1. `type`:`Literal["HumanMessageChunk"]`:= Stores the chunk-specific message type used during serialization and deserialization. Its default value is `"HumanMessageChunk"`.

````python
from langchain_core.messages import HumanMessageChunk


chunk1 = HumanMessageChunk(
    content="Explain ", # First partial human-message content
    name="student", # Inherited BaseMessage field passed through **kwargs
    id="human-chunk-101", # Inherited optional message identifier
    additional_kwargs={
        "source": "keyboard" # Provider-specific information
    },
    response_metadata={
        "chunk_number": 1 # Metadata associated with the first chunk
    }
)

chunk2 = HumanMessageChunk(
    content="SQL joins.", # Second partial human-message content
    name="student", # Same optional message name
    additional_kwargs={
        "language": "English" # Additional provider-specific information
    },
    response_metadata={
        "completed": True # Metadata associated with the second chunk
    }
)


combined_chunk = chunk1 + chunk2 # Combine the partial human-message chunks


print("Content:", combined_chunk.content) # Display merged message content
print("Type:", combined_chunk.type) # Display chunk-specific serialization type
print("Name:", combined_chunk.name) # Display inherited optional message name
print("ID:", combined_chunk.id) # Display inherited message identifier
print("Additional kwargs:", combined_chunk.additional_kwargs) # Display merged provider data
print("Response metadata:", combined_chunk.response_metadata) # Display merged metadata
print("Content blocks:", combined_chunk.content_blocks) # Display standardized content blocks
print("Text:", combined_chunk.text) # Display extracted textual content


print("\nPretty representation:")
print(
    combined_chunk.pretty_repr(
        html=False # Use plain-text formatting
    )
)


print("\nPretty print:")
combined_chunk.pretty_print() # Print the formatted human-message chunk
````
