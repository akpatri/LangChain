# Human Messages

The `human.py` module defines messages that represent input sent by a user to a chat model. It also provides a streaming chunk variant for partial human messages.

# HumanMessage

`HumanMessage` represents a message passed from a user to a chat model.

## Bases

- `BaseMessage`

## Attributes

1. `type`: Stores the message type used during serialization and deserialization.
   * **Type:**
     ```python
     type: Literal["human"] = "human"
     ```

### Methods

1. `__init__`: Initializes a human message using raw content or standardized content blocks.

   When `content_blocks` is provided, it is stored as the message content. Otherwise, the value supplied through `content` is used.

   * **Syntax:**
     ```python
     __init__(
         self,
         content: str | list[str | dict[Any, Any]] | None = None, # Raw message content
         content_blocks: list[types.ContentBlock] | None = None, # Standardized content blocks
         **kwargs: Any # Additional message fields
     ) -> None
     ```

# HumanMessageChunk

`HumanMessageChunk` represents a partial human message produced during streaming. It inherits message-chunk combination behavior from `BaseMessageChunk`.

## Bases

- `HumanMessage`
- `BaseMessageChunk`

## Attributes

1. `type`: Stores the chunk-specific message type used during serialization and deserialization.
   * **Type:**
     ```python
     type: Literal["HumanMessageChunk"] = "HumanMessageChunk"
     ```
