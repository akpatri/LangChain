# Model Profiles

The `model_profile.py` module defines the capability schema used by chat-model integrations to describe supported inputs, outputs, token limits, tool calling, structured output, and other model features.

Model profiles are a beta feature. The schema is a `total=False` `TypedDict`, so every field is optional and consumers should account for missing keys.

# ModelProfile

`ModelProfile` describes the known metadata and capabilities of a chat model.

Profile data may be generated from the LangChain model-profile data source and exposed through a chat model's `profile` attribute. The schema permits additional keys so provider packages can introduce newer profile fields without failing Pydantic validation.

## Bases

- `TypedDict`
- `total=False`

## Configuration

1. `__pydantic_config__`: Allows profile dictionaries to contain fields not currently declared by `ModelProfile`.
   * **Definition:**
     ```python
     __pydantic_config__ = ConfigDict(
         extra="allow"
     )
     ```

3. `release_date`: Stores the model release date in ISO 8601 format.

5. `open_weights`: Indicates whether the model's weights are openly available.
   * **Type:**

# ModelProfileRegistry

`ModelProfileRegistry` represents a mapping from model identifiers or names to their corresponding capability profiles.

## Type Alias

* **Definition:**
  ```python
  ModelProfileRegistry = dict[
      str,
      ModelProfile
  ]
  ```

## Schema Behaviour

- All `ModelProfile` fields are optional because the `TypedDict` uses `total=False`.
- Provider packages may include additional fields because the Pydantic configuration uses `extra="allow"`.
- The module contains an internal validation helper that compares supplied keys with the declared profile fields.
- Unknown keys produce a warning about a possible version mismatch between `langchain-core` and the provider package.
- If profile type hints cannot be resolved, the unknown-key check is skipped and the failure is written to debug logging.

## Internal Components Omitted

The following private implementation details are not documented as public API entries:

- `_warn_unknown_profile_keys`
- `logger`
