# LangChain Model Profile Reference

Developer-facing statements defined in `langchain_core.language_models.model_profile`.

# `ModelProfile: TypedDict(total=False)`

`ModelProfile` describes the metadata, limits, and capabilities of a chat model.

This is a beta schema. Every field is optional, so access values with `.get()` when a field may be absent.

## Model Metadata

```python
name: str # Human-readable model name
status: str # Model lifecycle status
release_date: str # Model release date in ISO 8601 format
last_updated: str # Last profile update date in ISO 8601 format
open_weights: bool # Whether model weights are openly available
```

## Input Capabilities

```python
max_input_tokens: int # Maximum number of input tokens
text_inputs: bool # Whether text input is supported
image_inputs: bool # Whether image input is supported
image_url_inputs: bool # Whether image URL input is supported
pdf_inputs: bool # Whether PDF input is supported
audio_inputs: bool # Whether audio input is supported
video_inputs: bool # Whether video input is supported
image_tool_message: bool # Whether ToolMessage content may contain images
pdf_tool_message: bool # Whether ToolMessage content may contain PDFs
```

## Output Capabilities

```python
max_output_tokens: int # Maximum number of output tokens
reasoning_output: bool # Whether reasoning output is supported
text_outputs: bool # Whether text output is supported
image_outputs: bool # Whether image output is supported
audio_outputs: bool # Whether audio output is supported
video_outputs: bool # Whether video output is supported
```

## Tool Calling

```python
tool_calling: bool # Whether tool calling is supported
tool_choice: bool # Whether explicit tool selection is supported
tool_call_streaming: bool # Whether streaming returns structured tool-call chunks
```

## Structured Output

```python
structured_output: bool # Whether native structured output is supported
```

## Other Capabilities

```python
attachment: bool # Whether file attachments are supported
temperature: bool # Whether a temperature parameter is supported
```

# `ModelProfileRegistry`

Maps model identifiers or names to their corresponding profiles.

```python
ModelProfileRegistry = dict[
    str, # Model identifier or model name
    ModelProfile, # Capability profile associated with the model
]
```

````python
from langchain_core.language_models.model_profile import ( # Import the model profile types
    ModelProfile, # Import the individual model profile schema
    ModelProfileRegistry, # Import the model profile registry type
) # Finish importing the profile types


demo_profile: ModelProfile = { # Create one model capability profile
    "name": "Demo Chat Model", # Set the human-readable model name
    "status": "active", # Set the model lifecycle status
    "release_date": "2026-01-10", # Set the model release date
    "last_updated": "2026-07-01", # Set the latest profile update date
    "open_weights": False, # Indicate that model weights are not openly available
    "max_input_tokens": 128000, # Set the maximum input-token limit
    "max_output_tokens": 8192, # Set the maximum output-token limit
    "text_inputs": True, # Indicate that text input is supported
    "image_inputs": True, # Indicate that image input is supported
    "pdf_inputs": True, # Indicate that PDF input is supported
    "text_outputs": True, # Indicate that text output is supported
    "reasoning_output": True, # Indicate that reasoning output is supported
    "tool_calling": True, # Indicate that tool calling is supported
    "tool_choice": True, # Indicate that explicit tool selection is supported
    "tool_call_streaming": True, # Indicate that tool calls can be streamed
    "structured_output": True, # Indicate that structured output is supported
    "temperature": True, # Indicate that temperature configuration is supported
} # Finish creating the profile


profile_registry: ModelProfileRegistry = { # Create a registry containing model profiles
    "demo-chat-model": demo_profile, # Map the model identifier to its profile
} # Finish creating the profile registry


selected_profile: ModelProfile = profile_registry["demo-chat-model"] # Retrieve one model profile


model_name: str = selected_profile.get("name", "Unknown model") # Safely read the optional model name

input_limit: int | None = selected_profile.get("max_input_tokens") # Safely read the input-token limit

supports_tools: bool = selected_profile.get("tool_calling", False) # Check whether tool calling is supported

supports_images: bool = selected_profile.get("image_inputs", False) # Check whether image input is supported

supports_audio: bool = selected_profile.get("audio_inputs", False) # Read an optional field that is absent


print("Model name:", model_name) # Display the model name

print("Maximum input tokens:", input_limit) # Display the input-token limit

print("Supports tool calling:", supports_tools) # Display tool-calling support

print("Supports image input:", supports_images) # Display image-input support

print("Supports audio input:", supports_audio) # Display the default value for the missing field
````
