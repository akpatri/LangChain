# Fake Embedding Models Reference

Developer-facing statements defined in `langchain_core.embeddings.fake`.

# `FakeEmbeddings: Embeddings, BaseModel`

Fake embedding model intended only for unit tests.

Each embedding is sampled independently from a normal distribution. It is a toy implementation and should not be used as a real embedding model.

## Fields

```python
size: int # Number of values in each embedding vector
```

## Constructor

```python
FakeEmbeddings(
    *,
    size: int, # Number of values in each embedding vector
) -> None
```

## Methods

### `embed_documents`

Generates one independent random embedding for each input text.

```python
embed_documents(
    self,
    texts: list[str], # Texts to embed
) -> list[list[float]] # Random embedding vector for each input text
```

The text values do not influence the generated vectors.

### `embed_query`

Generates an independent random embedding for a query.

```python
embed_query(
    self,
    text: str, # Query text
) -> list[float] # Random query embedding
```

The query text does not influence the generated vector.

---

# `DeterministicFakeEmbedding: Embeddings, BaseModel`

Deterministic fake embedding model intended only for unit tests.

The same text produces the same normally distributed embedding across calls. It is a toy implementation and should not be used as a real embedding model.

## Fields

```python
size: int # Number of values in each embedding vector
```

## Constructor

```python
DeterministicFakeEmbedding(
    *,
    size: int, # Number of values in each embedding vector
) -> None
```

## Methods

### `embed_documents`

Generates one deterministic embedding for each input text.

```python
embed_documents(
    self,
    texts: list[str], # Texts to embed
) -> list[list[float]] # Deterministic embedding vector for each input text
```

Repeated text values produce identical vectors.

### `embed_query`

Generates a deterministic embedding for a query.

```python
embed_query(
    self,
    text: str, # Query text
) -> list[float] # Deterministic query embedding
```

Calling this method repeatedly with the same text produces the same vector.

````python
#%pip install -U langchain-core#Install LangChain Core if it is not already installed

from langchain_core.embeddings.fake import DeterministicFakeEmbedding, FakeEmbeddings#Import both fake embedding models


documents = [#Create sample documents for testing
    "Python is a programming language.",#Create the first sample document
    "LangChain helps build LLM applications.",#Create the second sample document
    "Vector databases store embedding vectors.",#Create the third sample document
]#Finish the document list

query = "What stores embedding vectors?"#Create a sample search query


random_model = FakeEmbeddings(size=5)#Create a random fake embedding model with five dimensions
random_document_vectors = random_model.embed_documents(documents)#Generate random vectors for all documents
random_query_vector = random_model.embed_query(query)#Generate a random vector for the query


print("FakeEmbeddings results:")#Display the random-model heading
print(f"Document vectors: {random_document_vectors}")#Display all random document vectors
print(f"Query vector: {random_query_vector}")#Display the random query vector
print(f"Vector size: {len(random_query_vector)}")#Verify that the vector contains five values


first_random_vector = random_model.embed_query(query)#Embed the same query for the first time
second_random_vector = random_model.embed_query(query)#Embed the same query for the second time

print("\nSame query with FakeEmbeddings:")#Display the random-comparison heading
print(f"First vector: {first_random_vector}")#Display the first generated vector
print(f"Second vector: {second_random_vector}")#Display the second generated vector
print(f"Vectors are equal: {first_random_vector == second_random_vector}")#Show that random vectors normally differ


deterministic_model = DeterministicFakeEmbedding(size=5)#Create a deterministic fake embedding model
deterministic_document_vectors = deterministic_model.embed_documents(documents)#Generate deterministic document vectors
first_deterministic_vector = deterministic_model.embed_query(query)#Embed the query for the first time
second_deterministic_vector = deterministic_model.embed_query(query)#Embed the same query again


print("\nDeterministicFakeEmbedding results:")#Display the deterministic-model heading
print(f"Document vectors: {deterministic_document_vectors}")#Display the deterministic document vectors
print(f"First query vector: {first_deterministic_vector}")#Display the first deterministic query vector
print(f"Second query vector: {second_deterministic_vector}")#Display the second deterministic query vector
print(f"Vectors are equal: {first_deterministic_vector == second_deterministic_vector}")#Confirm that identical text produces identical vectors


assert len(random_document_vectors) == len(documents)#Verify that every document received an embedding
assert all(len(vector) == 5 for vector in random_document_vectors)#Verify that every random vector has five dimensions
assert len(deterministic_document_vectors) == len(documents)#Verify that every document received a deterministic embedding
assert first_deterministic_vector == second_deterministic_vector#Verify deterministic behaviour for unit testing

print("\nAll embedding tests passed successfully.")#Display the final test result
````
