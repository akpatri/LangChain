# Runnable

The `Runnable` class is the main abstract interface for executable LangChain components.

A Runnable accepts an input, performs an operation, and produces an output. It provides standard methods for synchronous execution, asynchronous execution, batching, streaming, schema inspection, configuration, retries, fallbacks, and Runnable composition.

# Import

The `Runnable` class can be imported from `langchain_core.runnables`.

**Syntax**

```python
from langchain_core.runnables import Runnable # Import the base Runnable class
```

# Class Definition

**Syntax**

```python
class Runnable(
    ABC, # Mark Runnable as an abstract base class
    Generic[Input, Output], # Define generic input and output types
):
    ... # Provide the base interface for executable LangChain components
```

````python
import asyncio
from operator import itemgetter
from pydantic import BaseModel
from langchain_core.runnables import RunnableLambda, RunnableParallel

double = RunnableLambda(lambda x: x * 2)

def multiply_by_two(x: int) -> int:
    return x * 2

typed_runnable = RunnableLambda(multiply_by_two)
````

## 15. `pipe`

Creates a sequence in which the current Runnable is followed by one or more Runnables.

The output of one Runnable becomes the input of the next Runnable.

**Syntax**

```python
pipe(
    self, # Runnable executed first
    *others: Runnable[Any, Any] | Callable[[Any], Any], # Runnables or callables executed afterward
    name: str | None = None, # Optional name assigned to the resulting sequence
) -> RunnableSerializable[Input, Other] # Return the composed Runnable sequence
```

````python
add_one = RunnableLambda(lambda x: x + 1)
square = RunnableLambda(lambda x: x**2)
chain = add_one.pipe(square)
print(chain.invoke(4))

chain = add_one | square
print(chain.invoke(20))
````

## 16. `pick`

Creates a Runnable that selects one or more keys from the dictionary output of the current Runnable.

**Syntax**

```python
pick(
    self, # Runnable whose output contains the required keys
    keys: str | list[str], # Single key or collection of keys to select
) -> RunnableSerializable[Any, Any] # Return a Runnable that selects the requested keys
```


````python
get_user = RunnableLambda(
    lambda _: {
        "name": "Saad",
        "age": 23,
        "city": "Delhi",
    }
)

name_runnable = get_user.pick("name")

print(name_runnable.invoke(None)) # Saad

details = get_user.pick(["name", "city"])
print(details.invoke(None)) # {'name': 'Saad', 'city': 'Delhi'}
````

## 17. `assign`

Creates a Runnable that adds new fields to the dictionary output of the current Runnable.

Each new field is calculated using a Runnable or callable.

**Syntax**


````python
student = RunnableLambda(
    lambda name: {
        "name": name,
        "marks": 80,
    }
)
with_result = student.assign(result=lambda data: "Pass" if data["marks"] >= 40 else "Fail")
print(with_result.invoke("Saad")) # {'name': 'Saad', 'marks': 80, 'result': 'Pass'}
````

## 20. `batch`

Processes multiple inputs synchronously.

The default implementation invokes the Runnable concurrently for the supplied inputs using a thread-pool executor.

**Syntax**

```python
batch(
    self, # Runnable instance
    inputs: list[Input], # Input values to process
    config: RunnableConfig | list[RunnableConfig] | None = None, # Shared or per-input runtime configuration
    *,
    return_exceptions: bool = False, # Return exceptions in the result list instead of raising them
    **kwargs: Any, # Additional arguments passed to every invocation
) -> list[Output] # Return one output or exception for each input
```

````python
results = double.batch([1, 2, 3, 4])
print(results)
# [2, 4, 6, 8]

#Using separate configurations:
results = double.batch(
    [1, 2],
    config=[
        {"tags": ["first"]},
        {"tags": ["second"]},
    ],
)
print(results) # [2, 4]

#Returning exceptions:
divide = RunnableLambda(lambda x: 10 / x)
results = divide.batch(
    [2, 0, 5],
    return_exceptions=True,
)
print(results) # [5.0, ZeroDivisionError(...), 2.0]
````

## 31. `bind`

Creates a new Runnable with fixed keyword arguments bound to the current Runnable.

**Syntax**

```python

    self, # Runnable instance
```

````python
from langchain_core.runnables import RunnableLambda  # Import RunnableLambda

def add_prefix(text: str, prefix: str) -> str:  # Define a function accepting a keyword argument
    return prefix + text  # Return the prefix joined with the input text

runnable = RunnableLambda(add_prefix)  # Convert the function into a Runnable

bound_runnable = runnable.bind(  # Create a new Runnable with a fixed argument
    prefix="Result: "  # Permanently bind the prefix argument
)

output = bound_runnable.invoke("Hello LangChain")  # Execute without passing prefix again

print(output)  # Display the final result
````

## 36. `with_retry`

Creates a new Runnable that automatically retries failed executions.

Retry behaviour can be controlled using exception types, exponential jitter, and the maximum number of attempts.

**Syntax**

```python
with_retry(
    self, # Runnable instance
    *,
    retry_if_exception_type: tuple[type[BaseException], ...] = (Exception,), # Exception types that trigger a retry
    wait_exponential_jitter: bool = True, # Apply exponential backoff with random jitter
    exponential_jitter_params: ExponentialJitterParams | None = None, # Optional exponential-jitter configuration
    stop_after_attempt: int = 3, # Maximum number of execution attempts
) -> Runnable[Input, Output] # Return a Runnable with retry behaviour
```

````python
attempts = {"count": 0}

def unstable_function(x):
    attempts["count"] += 1

    if attempts["count"] < 3:
        raise ValueError("Temporary failure")

    return x * 2

unstable = RunnableLambda(unstable_function)

retrying = unstable.with_retry(
    retry_if_exception_type=(ValueError,),
    stop_after_attempt=3,
)

print(retrying.invoke(5))
# 10
````

## 38. `with_fallbacks`

Creates a new Runnable that executes fallback Runnables when the current Runnable fails.

Fallbacks are attempted sequentially until one succeeds or every fallback fails.

**Syntax**

```python
with_fallbacks(
    self, # Primary Runnable executed first
    fallbacks: Sequence[Runnable[Input, Output]], # Alternative Runnables executed after failures
    *,
    exceptions_to_handle: tuple[type[BaseException], ...] = (Exception,), # Exception types that activate the fallbacks
    exception_key: str | None = None, # Optional input-dictionary key used to pass the previous exception
) -> RunnableWithFallbacksT[Input, Output] # Return a Runnable with fallback behaviour
```

````python
#single fallback
primary = RunnableLambda(
    lambda x: 10 / x
)

fallback = RunnableLambda(
    lambda x: "Could not divide"
)

safe_division = primary.with_fallbacks([fallback])

print(safe_division.invoke(2)) # 5.0

print(safe_division.invoke(0)) # Could not divide

#multiple fallback
primary = RunnableLambda(
    lambda x: (_ for _ in ()).throw(ValueError("Primary failed"))
)

fallback_one = RunnableLambda(
    lambda x: (_ for _ in ()).throw(ValueError("Fallback one failed"))
)

fallback_two = RunnableLambda(
    lambda x: f"Recovered: {x}"
)

runnable = primary.with_fallbacks(
    [fallback_one, fallback_two],
    exceptions_to_handle=(ValueError,),
)

print(runnable.invoke("data")) # Recovered: data

#Passing the exception to a fallback:
primary = RunnableLambda(
    lambda data: 10 / data["number"]
)

fallback = RunnableLambda(
    lambda data: f"Error: {data['error']}"
)

safe = primary.with_fallbacks(
    [fallback],
    exception_key="error",
)

print(safe.invoke({"number": 0})) # Error: division by zero
````
