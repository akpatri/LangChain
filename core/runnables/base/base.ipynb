# Developer-Facing Top-Level Statements

This document lists all developer-facing top-level declarations defined in `langchain_core.runnables.base`.

## Classes

### `Runnable`

Base abstraction for a unit of work that supports invocation, batching, streaming, transformation, and composition.

### `RunnableSerializable`

Serializable form of `Runnable` that can be represented using LangChain's serialization system.

### `RunnableSequence`

Executes multiple Runnables in sequence, passing the output of each Runnable to the next one.

### `RunnableParallel`

Executes multiple Runnables with the same input and returns their outputs together in a mapping.

### `RunnableGenerator`

Wraps a synchronous or asynchronous generator function as a streaming Runnable.

### `RunnableLambda`

Wraps a regular Python callable or asynchronous callable as a Runnable.

### `RunnableEachBase`

Base class for applying a bound Runnable independently to each item in an input list.

### `RunnableEach`

Applies a Runnable independently to every item in an input list.

### `RunnableBindingBase`

Base wrapper that delegates execution to another Runnable with bound arguments and configuration.

### `RunnableBinding`

Wraps another Runnable with fixed keyword arguments, configuration, types, or lifecycle listeners.

## Aliases and Type Aliases

### `RunnableMap`

Alias for `RunnableParallel`.

### `RunnableLike`

Type alias representing values that can be converted into a Runnable, including Runnables, callables, generators, and mappings.

## Functions

### `coerce_to_runnable()`

Converts a supported Runnable-like object into a Runnable.

### `chain()`

Decorator that converts a supported function or generator function into a Runnable.
