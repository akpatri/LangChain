# `RunnableEach`

`RunnableEach` applies one bound `Runnable` independently to every value in an input sequence and returns the results as a list.

## Inheritance

```python
class RunnableEach(RunnableEachBase[Input, Output]): # Apply one Runnable to multiple input values
```

## Type Parameters

```python
Input # Input type accepted by the bound Runnable
Output # Output type produced by the bound Runnable
```

## Fields

```python
bound: Runnable[Input, Output] # Runnable executed for each input value
name: str | None = None # Optional name used for tracing and debugging
```

## Constructor

```python
RunnableEach(
    *,
    bound: Runnable[Input, Output], # Runnable executed for every item
    name: str | None = None, # Optional custom Runnable name
) -> None # Initialize the RunnableEach instance
```

## Input and Output

```python
Sequence[Input] # Sequence supplied to RunnableEach
list[Output] # Results returned in input order
```

## Execution Behaviour

`invoke()` passes the input sequence to the bound Runnable's `batch()` method.

`ainvoke()` passes the input sequence to the bound Runnable's `abatch()` method.

A separate child callback configuration is created for each input value.

## Developer-Facing Properties and Methods

### `InputType`

Returns a list type containing the input type of the bound Runnable.

### `OutputType`

Returns a list type containing the output type of the bound Runnable.

### `get_input_schema`

Returns a schema representing a list of values accepted by the bound Runnable.

### `get_output_schema`

Returns a schema representing a list of values produced by the bound Runnable.

### `config_specs`

Returns the configurable-field specifications exposed by the bound Runnable.

### `get_graph`

Returns the execution graph of the bound Runnable.

### `is_lc_serializable`

Reports that `RunnableEach` is serializable by LangChain.

### `get_lc_namespace`

Returns the LangChain Runnable serialization namespace.

### `get_name`

Returns the explicitly supplied name or a generated name based on the bound Runnable.

The generated format is:

```text
RunnableEach<BoundRunnableName>
```

### `invoke`

Synchronously applies the bound Runnable to every input value and returns the results as a list.

### `ainvoke`

Asynchronously applies the bound Runnable to every input value and returns the results as a list.

### `bind`

Returns a new `RunnableEach` whose bound Runnable contains the supplied fixed keyword arguments.

### `with_config`

Returns a new `RunnableEach` whose bound Runnable contains the supplied configuration.

### `with_listeners`

Returns a new `RunnableEach` whose bound Runnable contains synchronous lifecycle listeners.

### `with_alisteners`

Returns a new `RunnableEach` whose bound Runnable contains asynchronous lifecycle listeners.

### `astream_events`

Not supported by `RunnableEach`.

Calling it raises `NotImplementedError` for event-stream versions `"v1"`, `"v2"`, and `"v3"`.

## Important Behaviour

- The same bound Runnable is used for every item.
- The returned list follows the order of the input sequence.
- Keyword arguments supplied to `invoke()` or `ainvoke()` are forwarded to the bound Runnable.
- Runtime configuration is applied to each individual execution.
- Empty input produces an empty output list through the bound batch operation.

````python
from langchain_core.runnables.base import RunnableEach, RunnableLambda # Import required Runnable classes

def square(number: int) -> int: # Define the operation applied to each value
    return number ** 2 # Return the square of the value

square_runnable = RunnableLambda(square) # Convert the function into a Runnable

each_runnable = RunnableEach( # Create a Runnable that processes a sequence
    bound=square_runnable # Apply the same Runnable to every sequence item
)

result = each_runnable.invoke([2, 3, 4]) # Process every number independently

print(result) # Display the list of results
````
