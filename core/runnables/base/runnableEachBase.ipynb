# `RunnableEachBase: RunnableSerializable[Sequence[Input], list[Output]]`

`RunnableEachBase` is a serializable base wrapper that applies one bound `Runnable` independently to every item in an input sequence.

Use it when creating a custom `RunnableEach` subclass with different constructor arguments.

## Type Parameters

```python
Input # Input type accepted by the bound Runnable
Output # Output type produced by the bound Runnable
```

## Field

```python
bound: Runnable[Input, Output] # Runnable applied to every input item
```

## Constructor

```python
RunnableEachBase(
    *,
    bound: Runnable[Input, Output], # Runnable applied independently to each input item
    name: str | None = None, # Optional name used for tracing and debugging
) -> None # Initialize the RunnableEachBase instance
```

## Input and Output

```python
Sequence[Input] # Sequence passed to RunnableEachBase
list[Output] # Results returned in the original input order
```

## Overridden Properties and Methods

### `InputType`

Returns a list type containing the input type of the bound `Runnable`.

### `get_input_schema`

Returns a schema representing a list of values accepted by the bound `Runnable`.

### `OutputType`

Returns a list type containing the output type of the bound `Runnable`.

### `get_output_schema`

Returns a schema representing a list of values produced by the bound `Runnable`.

### `config_specs`

Returns the configurable-field specifications exposed by the bound `Runnable`.

### `get_graph`

Returns the execution graph of the bound `Runnable`.

### `is_lc_serializable`

Returns `True`, indicating that the class supports LangChain serialization.

### `get_lc_namespace`

Returns the LangChain serialization namespace for Runnable objects.

### `invoke`

Synchronously applies the bound `Runnable` to every input item using batch execution.

### `ainvoke`

Asynchronously applies the bound `Runnable` to every input item using asynchronous batch execution.

### `astream_events`

Event streaming is not supported and raises `NotImplementedError`.

## Behaviour

- The same bound `Runnable` processes every item.
- Results are returned as a list in input order.
- Runtime configuration is propagated separately to each item.
- Additional keyword arguments are forwarded to the bound `Runnable`.

````python
from langchain_core.runnables import RunnableLambda # Import RunnableLambda
from langchain_core.runnables.base import RunnableEachBase # Import RunnableEachBase


def to_uppercase(text: str) -> str: # Define the operation applied to each item
    return text.upper() # Return the uppercase text


class UppercaseEach(RunnableEachBase[str, str]): # Create a custom RunnableEachBase subclass
    def __init__(self) -> None: # Define a simplified custom constructor
        super().__init__( # Initialize RunnableEachBase
            bound=RunnableLambda(to_uppercase) # Bind the operation applied to every item
        )


runnable = UppercaseEach() # Create the custom Runnable

result = runnable.invoke(["hello", "langchain"]) # Process all items independently

print(result) # Display the returned list
````
