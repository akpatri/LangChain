# Runnable Utility Functions and Types

The `utils.py` module provides shared utilities used by LangChain `Runnable` implementations.

It includes concurrency helpers, callable-signature inspection, coroutine context handling, Python AST visitors, callable-source analysis, addable stream aggregation, configurable-field definitions, and asynchronous-callable detection.

## Re-exports

1. `create_model`: Re-exports `langchain_core.utils.pydantic.create_model` for backward compatibility.

## Type Aliases

1. `Input`: Represents a contravariant Runnable input type.
   * **Definition:**
     ```python
     Input = TypeVar(
         "Input",
         contravariant=True
     )
     ```

2. `Output`: Represents a covariant Runnable output type.

   Runnable output values are generally expected to support concatenation or addition during streaming.

   * **Definition:**
     ```python
     Output = TypeVar(
         "Output",
         covariant=True
     )
     ```

3. `Addable`: Represents a type whose values satisfy the `SupportsAdd` protocol.
   * **Definition:**
     ```python
     Addable = TypeVar(
         "Addable",
         bound=SupportsAdd[Any, Any]
     )
     ```

4. `AnyConfigurableField`: Represents any supported configurable-field declaration.
   * **Definition:**
     ```python
     AnyConfigurableField = (
         ConfigurableField
         | ConfigurableFieldSingleOption
         | ConfigurableFieldMultiOption
     )
     ```

# IsLocalDict

`IsLocalDict` is an AST visitor that collects string keys accessed on a particular local dictionary variable.

It detects direct subscript access and calls to the dictionary's `get` method.

## Bases

- `ast.NodeVisitor`

## Attributes

1. `name`: Stores the local variable name being inspected.
   * **Type:**
     ```python
     name: str
     ```

2. `keys`: Stores the string dictionary keys discovered by the visitor.
   * **Type:**
     ```python
     keys: set[str]
     ```

### Methods

1. `__init__`: Initializes the visitor for one local dictionary name.
   * **Syntax:**
     ```python
     __init__(
         self,
         name: str, # Local dictionary variable name
         keys: set[str] # Set populated with detected keys
     ) -> None
     ```

2. `visit_Subscript`: Detects string-key subscript reads on the configured dictionary variable.
   * **Syntax:**
     ```python
     visit_Subscript(
         self,
         node: ast.Subscript # Subscript node to inspect
     ) -> None
     ```

3. `visit_Call`: Detects calls to the configured dictionary variable's `get` method when the first argument is a string constant.
   * **Syntax:**
     ```python
     visit_Call(
         self,
         node: ast.Call # Call node to inspect
     ) -> None
     ```

# NonLocals

`NonLocals` is an AST visitor that tracks loaded and stored names inside a callable's syntax tree.

It also records qualified attribute paths such as nested object attributes.

## Bases

- `ast.NodeVisitor`

## Attributes

1. `loads`: Stores names and qualified attribute paths that are read.
   * **Type:**
     ```python
     loads: set[str]
     ```

2. `stores`: Stores names that are assigned.
   * **Type:**
     ```python
     stores: set[str]
     ```

### Methods

1. `__init__`: Initializes empty sets for loaded and stored names.
   * **Syntax:**
     ```python
     __init__(
         self
     ) -> None
     ```

2. `visit_Name`: Records whether a name node is loaded or stored.
   * **Syntax:**
     ```python
     visit_Name(
         self,
         node: ast.Name # Name node to inspect
     ) -> None
     ```

3. `visit_Attribute`: Records qualified attribute expressions that are loaded.
   * **Syntax:**
     ```python
     visit_Attribute(
         self,
         node: ast.Attribute # Attribute node to inspect
     ) -> None
     ```

# AddableDict

`AddableDict` is a dictionary whose values can be combined through dictionary addition.

Keys found in only one dictionary are preserved. When both dictionaries contain a non-`None` value for the same key, their values are added when possible. If value addition raises `TypeError`, the value from the right-hand dictionary is used.

## Bases

- `dict[str, Any]`

### Methods

1. `__add__`: Adds another `AddableDict` to this dictionary and returns a new combined dictionary.
   * **Syntax:**
     ```python
     __add__(
         self,
         other: AddableDict # Dictionary whose values are added from the right
     ) -> AddableDict
     ```

2. `__radd__`: Adds this dictionary to another `AddableDict` and returns a new combined dictionary.
   * **Syntax:**
     ```python
     __radd__(
         self,
         other: AddableDict # Dictionary whose values are added from the left
     ) -> AddableDict
     ```

# SupportsAdd

`SupportsAdd` is a generic protocol for values that implement addition.

## Bases

### Methods

   * **Syntax:**

     ```

   * **Type:**

7. `dependencies`: Stores configuration-field identifiers on which this field depends.
   * **Type:**
     ```python
     dependencies: list[str] | None = None
     ```

### Functions

1. `gated_coro`: Runs one coroutine while holding an asynchronous semaphore.

   The semaphore limits how many gated coroutines may execute concurrently.

   * **Syntax:**
     ```python
     async gated_coro(
         semaphore: asyncio.Semaphore, # Semaphore controlling concurrent execution
         coro: Coroutine[
             Any,
             Any,
             Any
         ] # Coroutine to execute
     ) -> Any
     ```

2. `gather_with_concurrency`: Runs multiple coroutines and returns their results while optionally limiting concurrency.

   When `n` is `None`, the coroutines are passed directly to `asyncio.gather`. Otherwise, they are wrapped with one shared semaphore.

   * **Syntax:**
     ```python
     async gather_with_concurrency(
         n: int | None, # Maximum concurrent coroutine count
         *coros: Coroutine[
             Any,
             Any,
             Any
         ] # Coroutines to gather
     ) -> list[Any]
     ```

3. `accepts_run_manager`: Checks whether a callable's signature contains a parameter named `run_manager`.

   It returns `False` when the callable's signature cannot be inspected.

   * **Syntax:**
     ```python
     accepts_run_manager(
         callable: Callable[
             ...,
             Any
         ] # Callable to inspect
     ) -> bool
     ```

4. `accepts_config`: Checks whether a callable's signature contains a parameter named `config`.

   It returns `False` when the callable's signature cannot be inspected.

   * **Syntax:**
     ```python
     accepts_config(
         callable: Callable[
             ...,
             Any
         ] # Callable to inspect
     ) -> bool
     ```

5. `accepts_context`: Checks whether a callable's signature contains a parameter named `context`.

   It returns `False` when the callable's signature cannot be inspected.

   * **Syntax:**
     ```python
     accepts_context(
         callable: Callable[
             ...,
             Any
         ] # Callable to inspect
     ) -> bool
     ```

6. `asyncio_accepts_context`: Checks whether the running Python version supports the `context` parameter on `asyncio.create_task`.

   The function returns `True` for Python 3.11 and newer.

   * **Syntax:**
     ```python
     asyncio_accepts_context() -> bool
     ```

7. `coro_with_context`: Creates an asynchronous task that executes a coroutine within a supplied `Context`.

   On Python 3.11 and newer, the context is passed directly to `asyncio.create_task`. On older versions, task creation is executed through `Context.run`. The `create_task` parameter is retained for compatibility; the helper always creates a task.

   * **Syntax:**
     ```python
     coro_with_context(
         coro: Awaitable[_T], # Awaitable to execute
         context: Context, # Context used for task creation
         *,
         create_task: bool = False # Compatibility parameter
     ) -> Awaitable[_T]
     ```

8. `get_function_first_arg_dict_keys`: Extracts constant string keys accessed on the first argument of a function or lambda.

   It inspects direct subscript reads and calls to `get`. Detected keys are sorted. The function returns `None` when no keys are found or when the callable's source cannot be inspected or parsed.

   * **Syntax:**
     ```python
     get_function_first_arg_dict_keys(
         func: Callable[
             ...,
             Any
         ] # Callable whose first argument is inspected
     ) -> list[str] | None
     ```

9. `get_lambda_source`: Returns the source expression of a callable when exactly one lambda is found.

   When source extraction or parsing fails, or when the source contains a number of lambdas other than one, the callable's non-lambda name is returned when available.

   * **Syntax:**
     ```python
     get_lambda_source(
         func: Callable[
             ...,
             Any
         ] # Callable to inspect
     ) -> str | None
     ```

10. `get_function_nonlocals`: Returns the values of global and closure variables referenced by a callable.

    Qualified attribute references are resolved when possible. Results are cached for up to 256 callables. An empty list is returned when source or closure inspection fails.

    * **Syntax:**
      ```python
      get_function_nonlocals(
          func: Callable[
              ...,
              Any
          ] # Callable whose referenced values are inspected
      ) -> list[Any]
      ```

11. `indent_lines_after_first`: Indents every line after the first by the number of characters in `prefix`.

    The prefix's contents are not inserted; only its length determines the number of spaces.

    * **Syntax:**
      ```python
      indent_lines_after_first(
          text: str, # Multiline text to format
          prefix: str # String whose length determines indentation
      ) -> str
      ```

12. `add`: Adds an iterable of `SupportsAdd` values from left to right.

    The first item becomes the initial result. The function returns `None` for an empty iterable.

    * **Syntax:**
      ```python
      add(
          addables: Iterable[Addable] # Addable values to combine
      ) -> Addable | None
      ```

13. `aadd`: Asynchronously adds an `AsyncIterable` of `SupportsAdd` values from left to right.

    The first item becomes the initial result. The function returns `None` when the asynchronous iterable is empty.

    * **Syntax:**
      ```python
      async aadd(
          addables: AsyncIterable[Addable] # Asynchronous addable values to combine
      ) -> Addable | None
      ```

14. `get_unique_config_specs`: Sorts, groups, and deduplicates configurable-field specifications by identifier.

    Identical specifications sharing an identifier are reduced to one item. A `ValueError` is raised when the same identifier is associated with conflicting specifications.

    * **Syntax:**
      ```python
      get_unique_config_specs(
          specs: Iterable[
              ConfigurableFieldSpec
          ] # Configuration specifications to deduplicate
      ) -> list[
          ConfigurableFieldSpec
      ]
      ```

15. `is_async_generator`: Checks whether a function or callable object's `__call__` method is an asynchronous generator function.

    The return type is a type guard for callables producing `AsyncIterator[Any]`.

    * **Syntax:**
      ```python
      is_async_generator(
          func: Any # Function or callable object to inspect
      ) -> TypeGuard[
          Callable[
              ...,
              AsyncIterator[Any]
          ]
      ]
      ```

16. `is_async_callable`: Checks whether a function or callable object's `__call__` method is asynchronous.

    The return type is a type guard for callables producing `Awaitable[Any]`.

    * **Syntax:**
      ```python
      is_async_callable(
          func: Any # Function or callable object to inspect
      ) -> TypeGuard[
          Callable[
              ...,
              Awaitable[Any]
          ]
      ]
